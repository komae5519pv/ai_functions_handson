  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>2026 年 4〜9 月の注文明細（約 5,600 件）を作ります。<b>7〜9 月は「関西 × アプリ」の注文だけが大きく減る</b>ように作ってあります。これを関数が見つけられるかを確かめます。</div></div>

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.orders
COMMENT '注文明細（AI Functions デモ用の架空データ。7〜9月は関西×アプリの売上が落ちるように作ってある）'
AS
WITH g AS (
  SELECT
    id AS order_id,
    date_add(DATE'2026-04-01', CAST(abs(hash(id, 'd')) % 183 AS INT)) AS order_date,
    element_at(array('関東', '関西', '九州', '東北'), CAST(abs(hash(id, 'r')) % 4 AS INT) + 1) AS region,
    element_at(array('店舗', 'Web', 'アプリ'), CAST(abs(hash(id, 'c')) % 3 AS INT) + 1) AS channel,
    element_at(array('家電', 'オーディオ', 'キッチン家電', 'オフィス用品'), CAST(abs(hash(id, 'k')) % 4 AS INT) + 1) AS category,
    element_at(array('新規', 'リピーター'), CAST(abs(hash(id, 'u')) % 2 AS INT) + 1) AS customer_type,
    CAST(3000 + abs(hash(id, 'a')) % 27000 AS BIGINT) AS amount,
    abs(hash(id, 'x')) % 10 AS drop_key
  FROM range(6000)
)
SELECT order_id, order_date, region, channel, category, customer_type, amount
FROM g
-- 7月以降の「関西 × アプリ」は 70% の注文を間引く
WHERE NOT (order_date >= DATE'2026-07-01' AND region = '関西' AND channel = 'アプリ' AND drop_key < 7)
""")

# 列の説明（コメント）を付ける
column_comments = {
    'order_id'     : '注文ID',
    'order_date'   : '注文日',
    'region'       : 'お届け先の地域',
    'channel'      : '注文チャネル（店舗 / Web / アプリ）',
    'category'     : '商品カテゴリ',
    'customer_type': '会員区分（新規 / リピーター）',
    'amount'       : '注文金額（円）',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.orders ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.orders"))
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">4〜6月 と 7〜9月 の売上の差の要因を探す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>比較元（4〜6月）を FALSE、比較先（7〜9月）を TRUE にした列 <code>is_test</code> を作って渡します。</div></div>
</div>

```python
test_start = "2026-07-01"   # この日以降を比較先にする
dimensions = ["region", "channel", "category", "customer_type"]

df = spark.sql(f"""
WITH input AS (
  SELECT region, channel, category, customer_type, amount,
         order_date >= :test_start AS is_test
  FROM {CATALOG}.{SCHEMA}.orders
)
SELECT contributor, metric_control, metric_test, change,
       round(relative_change, 3) AS relative_change, round(support, 3) AS support
FROM ai_top_drivers(
  input       => TABLE(input),
  metric      => 'amount',
  is_test     => 'is_test',
  dimensions  => :dimensions,
  aggregation => 'sum'
)
ORDER BY abs(change) DESC
LIMIT 15
""", args={"test_start": test_start, "dimensions": dimensions})
display(df)
```

      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;"><code>["all"]</code> の行が全体の差です。その下に、差に大きく効いた切り口が並びます。仕込んだとおり <code>["region=関西", "channel=アプリ"]</code> の組み合わせが上位に来ているかを確認してください。単独の「関西」や「アプリ」より、組み合わせのほうが変化率（relative_change）が大きく出るはずです。</div>

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">件数（注文数）の差で見る</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>aggregation => 'count'</code> にすると件数の差を調べられます。数える列には、定数の列（<code>1 AS order_count</code>）を使うのが定番です。</div></div>

```python
df = spark.sql(f"""
WITH input AS (
  SELECT region, channel, category, customer_type, 1 AS order_count, order_date >= :test_start AS is_test
  FROM {CATALOG}.{SCHEMA}.orders
)
SELECT contributor, metric_control, metric_test, change, round(relative_change, 3) AS relative_change
FROM ai_top_drivers(input => TABLE(input), metric => 'order_count', is_test => 'is_test',
                    dimensions => :dimensions, aggregation => 'count')
ORDER BY abs(change) DESC
LIMIT 10
""", args={"test_start": test_start, "dimensions": dimensions})
display(df)
```

      <strong style="color: #e65100; font-size: 1.08em;">CREATE TABLE AS の中では、TABLE() にサブクエリを直接渡す</strong>
      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">作成時に試したところ、<code>CREATE TABLE ... AS</code> の中で WITH 句（CTE）の名前を <code>TABLE()</code> に渡すと内部エラーになりました（ふつうの SELECT では動きます）。保存するときは <code>TABLE(SELECT ...)</code> のようにサブクエリを直接渡します。</div>

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.sales_change_drivers
COMMENT 'orders の 4〜6月→7〜9月 の売上の差を ai_top_drivers で要因分解した結果'
AS
SELECT * FROM ai_top_drivers(
  input       => TABLE(SELECT region, channel, category, customer_type, amount, order_date >= :test_start AS is_test
                       FROM {CATALOG}.{SCHEMA}.orders),
  metric      => 'amount', is_test => 'is_test', dimensions => :dimensions, aggregation => 'sum')
""", args={"test_start": test_start, "dimensions": dimensions})

column_comments = {
    "contributor": "切り口（列=値 の配列。2つなら組み合わせ）",
    "change":      "比較先 − 比較元 の売上の差（円）",
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.sales_change_drivers ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.sales_change_drivers"))
```

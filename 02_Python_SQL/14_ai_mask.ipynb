```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.inquiries
COMMENT 'カスタマーサポートに届いた問い合わせ（AI Functions デモ用の架空データ）'
AS
SELECT * FROM VALUES
  (1, TIMESTAMP'2026-10-14 09:12:00', 'メール',      '注文番号 A-10231 の商品がまだ届きません。10月10日に注文し、到着予定は14日でした。明日のイベントで使うので至急確認してください。山田花子 090-1111-2222'),
  (2, TIMESTAMP'2026-10-14 10:40:00', 'Webフォーム', '請求書の金額が見積と違います。見積では税込132,000円でしたが、届いた請求書は145,200円になっています。確認をお願いします。経理部 佐藤 sato@example.co.jp'),
  (3, TIMESTAMP'2026-10-14 11:05:00', 'チャット',    'ポータブル電源PX-500を使っていたら焦げ臭いにおいがして、本体が異常に熱くなりました。怖くて使えません。どうすればいいですか。'),
  (4, TIMESTAMP'2026-10-14 13:22:00', 'Webフォーム', '会員登録したメールアドレスを変更したいです。新しいアドレスは k.tanaka@example.com です。手順を教えてください。'),
  (5, TIMESTAMP'2026-10-14 14:01:00', 'メール',      '先日購入したコーヒーメーカー、とても使いやすく満足しています。店頭スタッフの方の説明も丁寧でした。ありがとうございました。'),
  (6, TIMESTAMP'2026-10-14 15:30:00', 'メール',      '法人向けにノートPCを50台まとめて購入したいのですが、ボリュームディスカウントはありますか。来月中に導入したいです。株式会社サンプル 鈴木一郎 03-5555-1234'),
  (7, TIMESTAMP'2026-10-14 16:45:00', 'Webフォーム', '返品したいです。サイズが合いませんでした。未開封です。送り先は 大阪府大阪市北区梅田1-2-3 の高橋です。'),
  (8, TIMESTAMP'2026-10-14 18:10:00', 'チャット',    'アプリにログインできません。パスワードを再設定してもエラーになります。昨日から何度も試していて困っています。')
AS t(inquiry_id, received_at, channel, body)
""")

# 列の説明（コメント）を付ける
column_comments = {
    'inquiry_id' : '問い合わせID',
    'received_at': '受付日時',
    'channel'    : '受付チャネル（メール / Webフォーム / チャット）',
    'body'       : '問い合わせ本文（お客様が書いた原文。個人情報を含む）',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.inquiries ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries"))
```

```python
pii_types = ["person", "phone", "email", "address"]   # 伏せる情報の種類（Python のリスト）

df = spark.sql(
    "SELECT ai_mask(:text, :pii_types) AS masked",
    args={"text": "担当の山田太郎です。090-1234-5678 か taro.yamada@example.com までご連絡ください。東京都港区六本木1-2-3 に伺います。",
          "pii_types": pii_types})
display(df)
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">テーブルの列にまとめてかける</div><div style='color:#555; margin-top:6px; line-height:1.7;'>元の本文と、伏せた後の本文を並べて確認します。</div></div>

```python
df = spark.sql(f"""
SELECT
  inquiry_id,
  body,
  ai_mask(body, :pii_types) AS body_masked
FROM {CATALOG}.{SCHEMA}.inquiries
ORDER BY inquiry_id
""", args={"pii_types": pii_types})
display(df)
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">伏せる種類を変える</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>labels</code> に入れた種類だけが伏せられます。たとえば「個人の連絡先だけ伏せる」「会社名や注文番号も伏せる」のように、用途に合わせて選びます。</div></div>

```python
contact_only = ["phone", "email"]
names_and_orders = ["person", "organization", "order number"]

df = spark.sql(f"""
SELECT
  inquiry_id,
  body,
  ai_mask(body, :contact_only)     AS mask_contact_only,
  ai_mask(body, :names_and_orders) AS mask_names_and_orders
FROM {CATALOG}.{SCHEMA}.inquiries
WHERE inquiry_id IN (1, 2, 6)
ORDER BY inquiry_id
""", args={"contact_only": contact_only, "names_and_orders": names_and_orders})
display(df)
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.inquiries_masked
COMMENT 'inquiries の本文から ai_mask で個人情報を伏せた分析用テーブル'
AS
SELECT inquiry_id, received_at, channel, ai_mask(body, :pii_types) AS body_masked
FROM {CATALOG}.{SCHEMA}.inquiries
""", args={"pii_types": pii_types})

spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.inquiries_masked ALTER COLUMN body_masked COMMENT :c",
          args={"c": "氏名・電話番号・メールアドレス・住所を [MASKED] に置き換えた問い合わせ本文"})

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries_masked"))
```

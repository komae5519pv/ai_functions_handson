```python
import time
import pandas as pd
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.sql import StatementParameterListItem, StatementState

w = WorkspaceClient()

def run_on_warehouse(sql, **params):
    """SQL ウェアハウス（00_config の WAREHOUSE_ID）で SQL を実行し、結果を pandas の DataFrame で返す"""
    r = w.statement_execution.execute_statement(
        warehouse_id=WAREHOUSE_ID, catalog=CATALOG, schema=SCHEMA, statement=sql, wait_timeout="50s",
        parameters=[StatementParameterListItem(name=k, value=str(v)) for k, v in params.items()] or None)
    while r.status.state in (StatementState.PENDING, StatementState.RUNNING):
        time.sleep(3)
        r = w.statement_execution.get_statement(r.statement_id)
    if r.status.state != StatementState.SUCCEEDED:
        raise RuntimeError(r.status.error.message)
    if not (r.manifest and r.manifest.schema and r.manifest.schema.columns):
        return None
    cols = [c.name for c in r.manifest.schema.columns]
    return pd.DataFrame((r.result.data_array if r.result else None) or [], columns=cols)
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.daily_sales
COMMENT '店舗別の日次売上（AI Functions デモ用の架空データ。2025年4月〜2026年9月。右肩上がりで年末に山がある。異常値を意図的に3日分入れてある）'
AS
WITH d AS (
  SELECT explode(sequence(DATE'2025-04-01', DATE'2026-09-30')) AS sales_date
), s AS (
  SELECT * FROM VALUES ('新宿店', 1.00D), ('梅田店', 0.80D), ('博多店', 0.55D) AS s(store, scale)
), base AS (
  SELECT
    sales_date, store, scale,
    datediff(sales_date, DATE'2025-04-01') AS t,
    CASE WHEN dayofweek(sales_date) IN (1, 7) THEN 1.35D ELSE 1.0D END AS weekend,
    -- 季節の波: 年末（12/24 ごろ）に大きな山、夏（8/10 ごろ）に小さな山
    1.0D
      + 0.32D * exp(-pow(least(abs(dayofyear(sales_date) - 358), 365 - abs(dayofyear(sales_date) - 358)), 2) / (2 * 14 * 14))
      + 0.08D * exp(-pow(dayofyear(sales_date) - 222, 2) / (2 * 18 * 18)) AS season,
    (abs(hash(store, sales_date)) % 1000) / 1000.0D AS r
  FROM d CROSS JOIN s
)
SELECT
  sales_date,
  store,
  CAST(ROUND(
    2000000 * scale * weekend * season * (1 + t * 0.0015D) * (0.93D + r * 0.14D)
    * CASE
        WHEN store = '梅田店' AND sales_date = DATE'2026-08-15' THEN 0.35D   -- システム障害でレジ停止
        WHEN store = '新宿店' AND sales_date = DATE'2026-07-20' THEN 1.90D   -- 近隣イベントで急増
        WHEN store = '博多店' AND sales_date = DATE'2026-09-02' THEN 0.40D   -- 台風で臨時休業（半日）
        ELSE 1.0D
      END, -3) AS BIGINT) AS sales_amount,
  CAST(ROUND(480 * scale * weekend * (0.9D + r * 0.2D)) AS INT) AS visitors
FROM base
ORDER BY store, sales_date
""")

# 列の説明（コメント）を付ける
column_comments = {
    'sales_date'  : '営業日',
    'store'       : '店舗名',
    'sales_amount': '1日の売上金額（円）',
    'visitors'    : '1日の来店客数（人）',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.daily_sales ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.daily_sales"))
```

```python
horizon = "2026-12-31"      # 予測の期限（実績の最終日から 3 か月先まで）
holiday_region = "JP"

pdf = run_on_warehouse("""
SELECT *
FROM ai_forecast(
  TABLE(SELECT sales_date, store, sales_amount FROM daily_sales),
  horizon        => :horizon,
  time_col       => 'sales_date',
  value_col      => 'sales_amount',
  group_col      => 'store',
  holiday_region => :holiday_region,
  positive_only  => true,
  version        => '2'
)
ORDER BY store, sales_date
""", horizon=horizon, holiday_region=holiday_region)
forecast = spark.createDataFrame(pdf)
display(forecast)
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">実績と予測をつなげて見る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>実績（2026 年 4〜9 月）と予測（10〜12 月）を1つの表にして、1本の線でつながるグラフにします。学習には 1 年半分の実績をすべて使い（前の年の年末の山を覚えさせるため）、グラフには直近の半年だけを出します。グラフにしやすいよう、線の種類（実績・予測・下限・上限）を <code>series</code> 列に、値を <code>amount</code> 列に入れた縦長の表にし、予測の線は実績の最終日（9/30）から始めています。</div></div>

```python
show_from = "2026-04-01"   # グラフに出す実績の開始日
last_actual = "2026-09-30" # 実績の最終日（予測の線の始点にする）

pdf = run_on_warehouse("""
WITH fc AS (
  SELECT * FROM ai_forecast(
    TABLE(SELECT sales_date, store, sales_amount FROM daily_sales),
    horizon => :horizon, time_col => 'sales_date', value_col => 'sales_amount',
    group_col => 'store', holiday_region => :holiday_region, positive_only => true, version => '2')
),
actual AS (
  SELECT sales_date, store, CAST(sales_amount AS DOUBLE) AS amount
  FROM daily_sales
  WHERE sales_date >= CAST(:show_from AS DATE)
)
SELECT sales_date, store, '実績' AS series, amount FROM actual
UNION ALL
SELECT sales_date, store, s.series, amount
FROM actual CROSS JOIN VALUES ('予測'), ('下限'), ('上限') AS s(series)
WHERE sales_date = CAST(:last_actual AS DATE)
UNION ALL
SELECT sales_date, store,
  stack(3, '予測', sales_amount_forecast, '下限', sales_amount_lower, '上限', sales_amount_upper) AS (series, amount)
FROM fc
ORDER BY store, series, sales_date
""", horizon=horizon, holiday_region=holiday_region, show_from=show_from, last_actual=last_actual)
display(spark.createDataFrame(pdf))
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.campaign_sales
COMMENT '値引きキャンペーンの日がわかっている日次売上（AI Functions デモ用の架空データ。1店舗・2025年4月〜2026年11月。10〜11月は売上が空でキャンペーンの予定だけが入っている）'
AS
WITH d AS (
  SELECT explode(sequence(DATE'2025-04-01', DATE'2026-11-30')) AS sales_date
), b AS (
  SELECT
    sales_date,
    datediff(sales_date, DATE'2025-04-01') AS t,
    CASE WHEN dayofweek(sales_date) IN (1, 7) THEN 1.35D ELSE 1.0D END AS weekend,
    CASE
      -- 過去: 不定期に（およそ 10 日に 1 日）20% 引きのキャンペーンをした
      WHEN sales_date <= DATE'2026-09-30' AND abs(hash(sales_date, 'campaign')) % 10 = 0 THEN 0.2D
      -- これから: 10/10〜10/12 と 11/21〜11/23 に 20% 引きのキャンペーンを予定している
      WHEN sales_date IN (DATE'2026-10-10', DATE'2026-10-11', DATE'2026-10-12',
                          DATE'2026-11-21', DATE'2026-11-22', DATE'2026-11-23') THEN 0.2D
      ELSE 0.0D
    END AS discount_rate,
    (abs(hash('新宿店', sales_date)) % 1000) / 1000.0D AS r
  FROM d
)
SELECT
  sales_date,
  discount_rate,
  -- キャンペーンの日は売上が 1.6 倍になる。予測したい期間（10〜11月）は NULL
  CASE WHEN sales_date <= DATE'2026-09-30'
       THEN CAST(ROUND(2000000 * weekend * (1 + t * 0.0015D) * (0.93D + r * 0.14D) * (1 + discount_rate * 3), -3) AS BIGINT)
  END AS sales_amount
FROM b
ORDER BY sales_date
""")

# 列の説明（コメント）を付ける
column_comments = {
    'sales_date'   : '営業日',
    'discount_rate': 'その日の値引き率（0.2 = 20% 引きのキャンペーン。0 = キャンペーンなし）。2026年10〜11月は予定',
    'sales_amount' : '1日の売上金額（円）。予測したい期間（2026年10〜11月）は NULL',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.campaign_sales ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.campaign_sales"))
```

```python
cov_horizon = "2026-11-30"   # 予測の期限（campaign_sales の最後の日と同じ日にする）

pdf = run_on_warehouse("""
WITH with_campaign AS (
  -- 10〜11月の行（売上は NULL・値引き率は予定の値）も渡す
  SELECT * FROM ai_forecast(
    TABLE(SELECT sales_date, sales_amount, discount_rate FROM campaign_sales),
    horizon       => :cov_horizon,
    time_col      => 'sales_date',
    value_col     => 'sales_amount',
    covariate_col => 'discount_rate',
    positive_only => true,
    version       => '2')
), without_campaign AS (
  -- 比べるため: 実績の売上だけを渡す（キャンペーンの予定を知らない）
  SELECT * FROM ai_forecast(
    TABLE(SELECT sales_date, sales_amount FROM campaign_sales WHERE sales_amount IS NOT NULL),
    horizon       => :cov_horizon,
    time_col      => 'sales_date',
    value_col     => 'sales_amount',
    positive_only => true,
    version       => '2')
)
SELECT
  c.sales_date,
  c.discount_rate,
  round(n.sales_amount_forecast) AS forecast_without_campaign,
  round(w.sales_amount_forecast) AS forecast_with_campaign
FROM campaign_sales AS c
JOIN without_campaign AS n USING (sales_date)
JOIN with_campaign    AS w USING (sales_date)
WHERE c.sales_date BETWEEN DATE'2026-10-06' AND DATE'2026-10-15'
ORDER BY c.sales_date
""", cov_horizon=cov_horizon)
display(spark.createDataFrame(pdf))
```

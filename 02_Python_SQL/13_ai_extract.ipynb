```python
schema = '["注文番号", "製品名", "氏名", "電話番号", "メールアドレス"]'

df = spark.sql(f"""
SELECT
  inquiry_id,
  body,
  ai_extract(
    body,
    :schema,
    map('version', '2.1')
  ) AS result
FROM {CATALOG}.{SCHEMA}.inquiries
""", args={"schema": schema})
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">型と説明を付けて取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>項目ごとに <code>type</code> と <code>description</code> を付けると、数値は数値として、選択肢は決めた値の中から返ります。<code>enum</code> を使うと値を選択肢に限定できます。</div></div>

```python
schema = '''{
  "order_number":   {"type": "string",  "description": "注文番号（例: A-10231）。なければ null"},
  "amount_jpy":     {"type": "integer", "description": "文中に出てくる金額のうち、お客様が主張している金額（円）"},
  "quantity":       {"type": "integer", "description": "購入・注文したい数量"},
  "needs_reply_by": {"type": "string",  "description": "回答や対応の期限。文中に日付や「明日」などがあれば書く"},
  "customer_type":  {"type": "enum", "labels": ["個人", "法人"], "description": "問い合わせてきたのが個人か法人か"},
  "contact": {
    "type": "object",
    "properties": {
      "name":  {"type": "string", "description": "氏名"},
      "phone": {"type": "string", "description": "電話番号"},
      "email": {"type": "string", "description": "メールアドレス"}
    }
  }
}'''

df = spark.sql(f"""
SELECT
  inquiry_id,
  body,
  ai_extract(
    body,
    :schema,
    map('version', '2.1',
        'instructions', 'EC サイトのサポート窓口に届いた問い合わせです。書かれていない項目は推測せず null にしてください。')
  ) AS result
FROM {CATALOG}.{SCHEMA}.inquiries
""", args={"schema": schema})
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">確信度と引用を付ける</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>enableConfidenceScores</code> で項目ごとの確信度（0〜1）、<code>enableCitations</code> で原文のどの範囲から取ったか（<code>metadata.citations</code> の start / stop は文字位置）が返ります。人が確認すべき行を絞り込むのに使えます。</div></div>

```python
schema = '{"order_number": {"type": "string"}, "amount_jpy": {"type": "integer"}}'

df = spark.sql(f"""
SELECT
  inquiry_id,
  body,
  ai_extract(
    body,
    :schema,
    map('version', '2.1', 'enableConfidenceScores', 'true', 'enableCitations', 'true')
  ) AS result
FROM {CATALOG}.{SCHEMA}.inquiries
WHERE inquiry_id IN (1, 2)
""", args={"schema": schema})
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">6</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">VARIANT から列に取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>結果:response.項目.value::型</code> の形で列にします。入れ子の項目は <code>response.contact.name.value</code> のようにたどります。</div></div>
</div>

```python
schema = '''{
  "order_number": {"type": "string"},
  "amount_jpy":   {"type": "integer"},
  "contact": {"type": "object", "properties": {"name": {"type": "string"}, "phone": {"type": "string"}, "email": {"type": "string"}}}
}'''

df = spark.sql(f"""
WITH extracted AS (
  SELECT
    inquiry_id,
    ai_extract(
      body,
      :schema,
      map('version', '2.1')
    ) AS result
  FROM {CATALOG}.{SCHEMA}.inquiries
)
SELECT
  inquiry_id,
  result:response.order_number.value::string  AS order_number,
  result:response.amount_jpy.value::int       AS amount_jpy,
  result:response.contact.name.value::string  AS contact_name,
  result:response.contact.phone.value::string AS contact_phone,
  result:response.contact.email.value::string AS contact_email
FROM extracted
ORDER BY inquiry_id
""", args={"schema": schema})
display(df)
```

```python
schema = '''{
  "order_number": {"type": "string"},
  "amount_jpy":   {"type": "integer"},
  "contact": {"type": "object", "properties": {"name": {"type": "string"}, "phone": {"type": "string"}, "email": {"type": "string"}}}
}'''

spark.sql(f"""
CREATE OR REPLACE TABLE inquiries_extracted
COMMENT 'inquiries から ai_extract で注文番号・金額・連絡先を取り出したテーブル'
AS
WITH extracted AS (
  SELECT
    *,
    ai_extract(
      body,
      :schema,
      map('version', '2.1')
    ) AS result
  FROM {CATALOG}.{SCHEMA}.inquiries
)
SELECT
  inquiry_id, received_at, channel, body,
  result:response.order_number.value::string  AS order_number,
  result:response.amount_jpy.value::int       AS amount_jpy,
  result:response.contact.name.value::string  AS contact_name,
  result:response.contact.phone.value::string AS contact_phone,
  result:response.contact.email.value::string AS contact_email
FROM extracted
""", args={"schema": schema})

# 列の説明（コメント）を付ける
column_comments = {
    'order_number' : 'AI が本文から取り出した注文番号',
    'amount_jpy'   : 'AI が本文から取り出した金額（円）',
    'contact_name' : 'AI が本文から取り出した氏名',
    'contact_phone': 'AI が本文から取り出した電話番号',
    'contact_email': 'AI が本文から取り出したメールアドレス',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.inquiries_extracted ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

df = spark.sql(f"""
SELECT * FROM inquiries_extracted
""")
display(df)
```

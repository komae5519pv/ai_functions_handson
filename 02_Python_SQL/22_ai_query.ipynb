```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.reviews
COMMENT 'ECサイトの商品レビュー（AI Functions デモ用の架空データ）'
AS
SELECT * FROM VALUES
  (1, 'ワイヤレスイヤホン AirTone S2',         5, '軽くて耳が痛くならない。バッテリーも一日もつので出張に最適です。'),
  (2, 'ロボット掃除機 CleanBot X',             1, '2週間で電源が入らなくなった。サポートに連絡しても返事が遅く最悪です。'),
  (3, 'ワイヤレスイヤホン AirTone S2',         3, '音質は良いが、ケースが安っぽく傷がつきやすい。値段相応かな。'),
  (4, '全自動コーヒーメーカー BrewMaster Pro', 3, '説明書がわかりにくく設定に1時間かかった。動き出してからは毎朝快適です。'),
  (5, 'ポータブル電源 PX-500',                 4, '期待通りの性能。キャンプで2日使えた。配送も早かった。'),
  (6, 'ノートPCスタンド FlexLift',             2, '色が写真と全然違う。ぐらつきもあるので返品を考えています。'),
  (7, '全自動コーヒーメーカー BrewMaster Pro', 5, '毎日使っています。豆から挽けるので香りが段違い。もう手放せません！'),
  (8, 'ロボット掃除機 CleanBot X',             3, '普通です。特に良くも悪くもない。')
AS t(review_id, product_name, rating, review_text)
""")

# 列の説明（コメント）を付ける
column_comments = {
    'review_id'   : 'レビューID',
    'product_name': 'レビュー対象の商品名',
    'rating'      : '星の数（1〜5）',
    'review_text' : 'レビュー本文（お客様の原文）',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.reviews ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.reviews"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">出力の形を JSON に固定する（responseFormat）</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>responseFormat</code> で型を指定すると、決まった形の JSON が返るので、後続の処理で列に取り出しやすくなります。</div></div>

<div style="border-left: 4px solid #ff9800; background: #fff3e0; padding: 14px 20px; border-radius: 4px; margin: 14px 0;">

      <strong style="color: #e65100; font-size: 1.08em;">responseFormat に対応していないモデルがある</strong>
      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">作成時に試したところ、<code>databricks-claude-haiku-5-5</code> では「Structured output is not supported for this model」というエラーになりました（<code>databricks-claude-sonnet-4-6</code> では動作）。エラーになったら <code>LLM_ENDPOINT</code> を変えてください。</div>

```python
response_format = "STRUCT<result: STRUCT<good_points: ARRAY<STRING>, bad_points: ARRAY<STRING>, reply_needed: BOOLEAN>>"

df = spark.sql(f"""
SELECT
  review_id,
  review_text,
  ai_query(
    :llm_endpoint,
    concat('次の商品レビューを分析してください。\\nレビュー: ', review_text),
    responseFormat => :response_format
  ) AS analysis
FROM {CATALOG}.{SCHEMA}.reviews
""", args={"llm_endpoint": LLM_ENDPOINT, "response_format": response_format})
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">JSON を列に取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>返ってくるのは JSON の文字列なので、<code>from_json</code> で型を付けて列にします。</div></div>
</div>

```python
analysis_schema = "good_points ARRAY<STRING>, bad_points ARRAY<STRING>, reply_needed BOOLEAN"   # from_json で付ける型

df = spark.sql(f"""
WITH analyzed AS (
  SELECT
    review_id,
    review_text,
    ai_query(:llm_endpoint,
             concat('次の商品レビューを分析してください。\\nレビュー: ', review_text),
             responseFormat => :response_format) AS analysis
  FROM {CATALOG}.{SCHEMA}.reviews
)
SELECT review_id, review_text, a.good_points, a.bad_points, a.reply_needed
FROM (SELECT *, from_json(analysis, :analysis_schema) AS a FROM analyzed)
""", args={"llm_endpoint": LLM_ENDPOINT, "response_format": response_format, "analysis_schema": analysis_schema})
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">6</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">エラーで止めない・生成の設定を変える</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>failOnError => false</code> にすると、一部の行でエラーが起きてもクエリは最後まで動き、エラーは <code>errorMessage</code> に入ります。大量の行を処理するバッチでは付けておくと安心です。<code>modelParameters</code> で温度（ばらつき）や最大トークン数を変えられます。</div></div>
</div>

```python
temperature, max_tokens = 0.2, 200   # 生成の設定

df = spark.sql(f"""
SELECT review_id, review_text, r.result AS reply_draft, r.errorMessage AS error_message
FROM (
  SELECT
    review_id,
    review_text,
    ai_query(
      :llm_endpoint,
      concat('次のレビューへの、店舗からの返信を 60 文字以内で1つ書いてください。\\nレビュー: ', review_text),
      failOnError => false,
      modelParameters => named_struct('temperature', {temperature}, 'max_tokens', {max_tokens})
    ) AS r
  FROM {CATALOG}.{SCHEMA}.reviews
  WHERE rating <= 3
)
""", args={"llm_endpoint": LLM_ENDPOINT})
display(df)
```

```python
receipt_format = "STRUCT<result: STRUCT<store: STRING, date: STRING, total_jpy: INT>>"
images_path = f"{VOL_ROOT}/images/"

df = spark.sql("""
SELECT
  path,
  ai_query(:llm_endpoint, 'このレシートの店名・日付・合計金額を読み取ってください。',
           files => content, responseFormat => :receipt_format) AS receipt
FROM read_files(:images_path, format => 'binaryFile')
""", args={"llm_endpoint": LLM_ENDPOINT, "receipt_format": receipt_format, "images_path": images_path})
display(df)
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.reviews_analyzed
COMMENT 'reviews を ai_query で分析し、良い点・悪い点・返信要否を付けたテーブル'
AS
WITH analyzed AS (
  SELECT *,
         ai_query(:llm_endpoint,
                  concat('次の商品レビューを分析してください。\\nレビュー: ', review_text),
                  responseFormat => :response_format) AS analysis
  FROM {CATALOG}.{SCHEMA}.reviews
)
SELECT review_id, product_name, rating, review_text, a.good_points, a.bad_points, a.reply_needed
FROM (SELECT *, from_json(analysis, :analysis_schema) AS a FROM analyzed)
""", args={"llm_endpoint": LLM_ENDPOINT, "response_format": response_format, "analysis_schema": analysis_schema})

column_comments = {
    "good_points":  "AI がレビューから取り出した良い点",
    "bad_points":   "AI がレビューから取り出した悪い点",
    "reply_needed": "店舗から返信すべきレビューか",
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.reviews_analyzed ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.reviews_analyzed"))
```

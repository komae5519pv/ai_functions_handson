  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>誤字・脱字を含む日本語と英語の文章を 6 件作ります。</div></div>

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.draft_texts
COMMENT '誤字・脱字・文法の誤りを含む下書きの文章（AI Functions デモ用の架空データ）'
AS
SELECT * FROM VALUES
  (1, '商品説明',   'この商品わ、軽くて持ち運びやすいので、出張や旅行お供に最適です。'),
  (2, '商品説明',   '最大30時間の再生が可能で、一日中音楽お楽しめますす。'),
  (3, '問い合わせ', '先週注文した商品がとどいてないので、確認してもらえますでしょか。'),
  (4, '社内メモ',   '明日の会議わ10時からに変更になりました、資料わ各自で印刷おねがいします'),
  (5, 'English',    'This product have a long battery life and it are very light.'),
  (6, 'English',    'We was very happy with the fast delivery, thank you so much!')
AS t(text_id, text_type, draft)
""")

# 列の説明（コメント）を付ける
column_comments = {
    'text_id'  : '文章ID',
    'text_type': '文章の種類（商品説明 / 問い合わせ / 社内メモ / English）',
    'draft'    : '誤りを含む下書きの文章',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.draft_texts ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.draft_texts"))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">まとめて直す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>元の文と直した文を並べて確認します。</div></div>
</div>

```python
df = spark.sql(f"""
SELECT
  text_id,
  text_type,
  draft,
  ai_fix_grammar(draft) AS fixed
FROM {CATALOG}.{SCHEMA}.draft_texts
ORDER BY text_id
""")
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">どこが変わったかを確かめる</div><div style='color:#555; margin-top:6px; line-height:1.7;'>直す前後で文字列が変わった行だけを抜き出すと、確認する量を減らせます。</div></div>
</div>

```python
df = spark.sql(f"""
SELECT text_id, draft, fixed
FROM (
  SELECT text_id, draft, ai_fix_grammar(draft) AS fixed
  FROM {CATALOG}.{SCHEMA}.draft_texts
)
WHERE draft <> fixed
ORDER BY text_id
""")
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div></div>
</div>

```python
spark.sql(f"""
CREATE OR REPLACE TABLE draft_texts_fixed
COMMENT 'draft_texts を ai_fix_grammar で校正したテーブル'
AS
SELECT *, ai_fix_grammar(draft) AS fixed
FROM {CATALOG}.{SCHEMA}.draft_texts
""")

# 列の説明（コメント）を付ける
column_comments = {
    'fixed': 'AI が誤字・脱字・文法を直した文章',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.draft_texts_fixed ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

df = spark.sql(f"""
SELECT * FROM draft_texts_fixed
""")
display(df)
```

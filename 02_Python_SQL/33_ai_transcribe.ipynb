```python
df = spark.sql(f"""
SELECT
  element_at(split(path, '/'), -1) AS file_name,
  ai_transcribe(content)        AS transcription
FROM read_files(:path, format => 'binaryFile')
""", args={"path": f"{VOL_ROOT}/audio/"})
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">発話ごとに1行にする</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>segments</code> を行に展開すると、誰がいつ何を話したかの表になります。</div></div>
</div>

```python
df = spark.sql(f"""
WITH t AS (
  SELECT element_at(split(path, '/'), -1) AS file_name, ai_transcribe(content) AS result
  FROM read_files(:path, format => 'binaryFile')
)
SELECT
  t.file_name,
  round(seg.value:start::double, 1) AS start_sec,
  round(seg.value:end::double, 1)   AS end_sec,
  seg.value:speaker_id::string      AS speaker,
  seg.value:text::string            AS text
FROM t,
LATERAL variant_explode(t.result:response.segments) AS seg
ORDER BY file_name, start_sec
""", args={"path": f"{VOL_ROOT}/audio/"})
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">全文にして、ほかの AI 関数につなげる</div><div style='color:#555; margin-top:6px; line-height:1.7;'>発話を順につなげて全文にし、<code>ai_extract</code>（項目の取り出し）と <code>ai_analyze_sentiment</code>（感情）にかけます。通話ごとの「困りごと・案内した解決策・後日対応の要否」が表になります。</div></div>
</div>

```python
value = '[話者'
schema = '''{
  "product":        {"type": "string",  "description": "話題になっている製品"},
  "issue":          {"type": "string",  "description": "お客様の困りごと（日本語で）"},
  "resolution":     {"type": "string",  "description": "オペレーターが案内した解決策（日本語で）"},
  "order_number":   {"type": "string",  "description": "注文番号。言っていなければ null"},
  "follow_up_needed": {"type": "boolean", "description": "後日の対応が必要か"}
}'''

df = spark.sql(f"""
WITH t AS (
  SELECT element_at(split(path, '/'), -1) AS file_name, ai_transcribe(content) AS result
  FROM read_files(:path, format => 'binaryFile')
),
full_text AS (
  SELECT
    file_name,
    array_join(
      transform(cast(result:response.segments AS ARRAY<VARIANT>),
                s -> concat(:value, s:speaker_id::string, '] ', s:text::string)),
      '\n') AS transcript
  FROM t
)
SELECT
  file_name,
  transcript,
  ai_analyze_sentiment(transcript) AS sentiment,
  ai_extract(
    transcript,
    :schema,
    map('version', '2.1')
  ):response AS call_summary
FROM full_text
""", args={"value": value, "schema": schema, "path": f"{VOL_ROOT}/audio/"})
display(df)
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div></div>
</div>

```python
spark.sql(f"""
CREATE OR REPLACE TABLE call_transcripts
COMMENT '通話録音を ai_transcribe で文字に起こした結果（発話ごと）'
AS
WITH t AS (
  SELECT element_at(split(path, '/'), -1) AS file_name, ai_transcribe(content) AS result
  FROM read_files(:path, format => 'binaryFile')
)
SELECT
  t.file_name,
  seg.value:start::double      AS start_sec,
  seg.value:end::double        AS end_sec,
  seg.value:speaker_id::string AS speaker,
  seg.value:text::string       AS text
FROM t,
LATERAL variant_explode(t.result:response.segments) AS seg
""", args={"path": f"{VOL_ROOT}/audio/"})

# 列の説明（コメント）を付ける
column_comments = {
    'file_name': '録音ファイル名',
    'start_sec': '発話の開始（秒）',
    'end_sec'  : '発話の終了（秒）',
    'speaker'  : '話者ID（話し始めた順に 1, 2, ...）',
    'text'     : '書き起こしたテキスト',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.call_transcripts ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

df = spark.sql(f"""
SELECT * FROM call_transcripts ORDER BY file_name, start_sec
""")
display(df)
```

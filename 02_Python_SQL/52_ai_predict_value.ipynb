  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">2</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>既存店 140 店の条件と月商、出店候補地 10 か所の条件を作ります。候補地は <code>monthly_sales</code> が NULL です。</div></div>
</div>

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.store_locations
COMMENT '既存店の実績と出店候補地の条件（AI Functions デモ用の架空データ）'
AS
WITH f AS (
  SELECT
    id + 1 AS location_id,
    element_at(array('駅前', '郊外', '商業施設内'), CAST(abs(hash(id, 'a')) % 3 AS INT) + 1) AS area_type,
    CAST(80 + abs(hash(id, 'f')) % 420 AS INT) AS floor_area_m2,
    CAST(1 + abs(hash(id, 'w')) % 25 AS INT) AS walk_min_from_station,
    CAST(abs(hash(id, 'p')) % 120 AS INT) AS parking_spaces,
    CAST(3000 + abs(hash(id, 'o')) % 45000 AS INT) AS population_1km,
    0.9D + (abs(hash(id + 1, 'n')) % 200) / 1000.0D AS noise
  FROM range(150)
)
SELECT
  location_id, area_type, floor_area_m2, walk_min_from_station, parking_spaces, population_1km,
  CASE WHEN location_id > 140 THEN NULL   -- 最後の10か所は出店前（＝予測対象）
       ELSE CAST(ROUND((floor_area_m2 * 38000
              + population_1km * 120
              - walk_min_from_station * 250000
              + parking_spaces * 30000
              + CASE area_type WHEN '駅前' THEN 4000000 WHEN '商業施設内' THEN 2500000 ELSE 0 END) * noise, -4) AS BIGINT)
  END AS monthly_sales
FROM f
""")

# 列の説明（コメント）を付ける
column_comments = {
    'location_id'          : '店舗・候補地ID',
    'area_type'            : '立地タイプ（駅前 / 郊外 / 商業施設内）',
    'floor_area_m2'        : '売場面積（平方メートル）',
    'walk_min_from_station': '最寄り駅からの徒歩分数',
    'parking_spaces'       : '駐車場の台数',
    'population_1km'       : '半径1km の人口（人）',
    'monthly_sales'        : '月商（円）。NULL はまだ出店していない候補地＝予測したい行',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.store_locations ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.store_locations"))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">候補地の月商を予測する</div></div>
</div>

```python
target_col = "monthly_sales"
feature_cols = ["area_type", "floor_area_m2", "walk_min_from_station", "parking_spaces", "population_1km"]
seed = 42

df = spark.sql(f"""
SELECT * FROM ai_predict_value(
  input        => TABLE({CATALOG}.{SCHEMA}.store_locations),
  target_col   => :target_col,
  feature_cols => :feature_cols,
  seed         => :seed
)
ORDER BY monthly_sales_prediction DESC
""", args={"target_col": target_col, "feature_cols": feature_cols, "seed": seed})
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">どのくらい当たるかを確かめる</div><div style='color:#555; margin-top:6px; line-height:1.7;'>既存店の一部（ID が 5 の倍数）の月商をわざと隠して予測させ、実際の月商との誤差（平均絶対誤差・平均誤差率）を出します。</div></div>
</div>

```python
df = spark.sql(f"""
WITH holdout AS (
  SELECT *, monthly_sales AS actual, CASE WHEN location_id % 5 = 0 THEN NULL ELSE monthly_sales END AS sales_masked
  FROM {CATALOG}.{SCHEMA}.store_locations
  WHERE monthly_sales IS NOT NULL
),
predicted AS (
  SELECT * FROM ai_predict_value(
    input        => TABLE(SELECT * EXCEPT (monthly_sales) FROM holdout),
    target_col   => 'sales_masked',
    feature_cols => :feature_cols,
    seed         => :seed
  )
)
SELECT count(*) AS test_rows,
       round(avg(abs(sales_masked_prediction - actual)), -3)         AS mean_abs_error_jpy,
       round(avg(abs(sales_masked_prediction - actual) / actual), 3) AS mean_abs_pct_error
FROM predicted
""", args={"feature_cols": feature_cols, "seed": seed})
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">予測結果をテーブルに保存する</div></div>
</div>

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.store_sales_prediction
COMMENT 'store_locations の出店候補地について ai_predict_value で月商を予測した結果'
AS
SELECT location_id, area_type, monthly_sales_prediction
FROM ai_predict_value(input => TABLE({CATALOG}.{SCHEMA}.store_locations), target_col => :target_col,
                      feature_cols => :feature_cols, seed => :seed)
""", args={"target_col": target_col, "feature_cols": feature_cols, "seed": seed})

spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.store_sales_prediction ALTER COLUMN monthly_sales_prediction COMMENT :c",
          args={"c": "予測した月商（円）"})

display(spark.table(f"{CATALOG}.{SCHEMA}.store_sales_prediction"))
```

```python
import json

question = "返品するとき、送料は誰が払うの？"
# 情報源の設定（Python の辞書で書いて JSON にする）
knowledge_sources = json.dumps([{
    "type": "vector_search",
    "config": {"index_name": VS_INDEX, "text_col": "body", "doc_uri_col": "url"},
}])

df = spark.sql("SELECT ai_search(:question, PARSE_JSON(:knowledge_sources)) AS result",
               args={"question": question, "knowledge_sources": knowledge_sources})
display(df)
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">答えと出典を列に取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>結果:answer::string</code> で答え、<code>結果:document[0].doc_uri</code> で一番関連の高い文書の URL を取り出します。<code>generate_citations</code> を付けると、答えの根拠になった文書の番号（<code>citations</code>）も返ります。</div></div>

```python
df = spark.sql("""
WITH r AS (
  SELECT ai_search(:question, PARSE_JSON(:knowledge_sources), options => map('generate_citations', 'true')) AS result
)
SELECT
  result:answer::string              AS answer,
  result:document[0].doc_uri::string AS top_source,
  result:citations                   AS citations,
  result:document                    AS documents
FROM r
""", args={"question": "ポータブル電源が熱くなって変なにおいがする。どうしたらいい？", "knowledge_sources": knowledge_sources})
display(df)
```

```python
instructions = "カスタマーサポートの FAQ から、お客様の困りごとに直接答える記事を探してください。安全に関わる内容を最優先してください。"

df = spark.sql(f"""
SELECT inquiry_id, body, r:answer::string AS suggested_reply, r:document[0].doc_uri::string AS source_url
FROM (
  SELECT inquiry_id, body, ai_search(body, PARSE_JSON(:knowledge_sources), :instructions) AS r
  FROM {CATALOG}.{SCHEMA}.inquiries
)
ORDER BY inquiry_id
""", args={"knowledge_sources": knowledge_sources, "instructions": instructions})
display(df)
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.inquiries_suggested_reply
COMMENT 'inquiries の各問い合わせに、ai_search で FAQ を根拠にした回答案を付けたテーブル'
AS
SELECT inquiry_id, body, r:answer::string AS suggested_reply, r:document[0].doc_uri::string AS source_url
FROM (
  SELECT inquiry_id, body, ai_search(body, PARSE_JSON(:knowledge_sources)) AS r
  FROM {CATALOG}.{SCHEMA}.inquiries
)
""", args={"knowledge_sources": knowledge_sources})

column_comments = {
    "suggested_reply": "FAQ をもとに AI が作った回答の下書き",
    "source_url":      "回答の根拠として最も関連の高かった FAQ 記事の URL",
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.inquiries_suggested_reply ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries_suggested_reply"))
```

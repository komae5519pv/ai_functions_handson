  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>自社の商品マスタ（products）と、表記の違う仕入先の納品データ（supplier_items）を作ります。</div></div>

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.products
COMMENT '自社の商品マスタ（AI Functions デモ用の架空データ）'
AS
SELECT * FROM VALUES
  ('P001', 'ポータブル電源 PX-500',                 '家電',         '容量512Wh / AC出力500W / 重量6.2kg / ソーラー充電対応', 69800),
  ('P002', 'ワイヤレスイヤホン AirTone S2',         'オーディオ',   'ノイズキャンセリング / 最大30時間再生 / 防水IPX4',      14800),
  ('P003', '電動アシスト自転車 RideOn 26',          'モビリティ',   '26インチ / 航続距離80km / 7段変速',                     128000),
  ('P004', '全自動コーヒーメーカー BrewMaster Pro', 'キッチン家電', 'ミル付き / 最大10杯 / タイマー予約',                    32800),
  ('P005', 'ノートPCスタンド FlexLift',             'オフィス用品', 'アルミ製 / 高さ6段階 / 折りたたみ式',                   4980),
  ('P006', 'ロボット掃除機 CleanBot X',             '生活家電',     'マッピング機能 / 自動ゴミ収集 / 水拭き対応',            59800)
AS t(product_id, product_name, category, spec, price_jpy)
""")

# 列の説明（コメント）を付ける
column_comments = {
    'product_id'  : '商品ID',
    'product_name': '自社の正式な商品名',
    'category'    : '商品カテゴリ',
    'spec'        : '主な仕様（社内の商品マスタに登録された短いメモ）',
    'price_jpy'   : '税込販売価格（円）',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.products ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.products"))
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.{SCHEMA}.supplier_items
COMMENT '仕入先から届いた納品データ。自社の商品マスタとは表記が違う（AI Functions デモ用の架空データ）'
AS
SELECT * FROM VALUES
  ('S-901', 'PX500 ポータブル電源（512Wh）'),
  ('S-902', 'AirTone S2 ワイヤレス イヤホン 黒'),
  ('S-903', 'コーヒーメーカー BrewMaster（ミル付）'),
  ('S-904', 'CleanBot-X ロボットクリーナー'),
  ('S-905', '電動自転車 ライドオン 26インチ'),
  ('S-906', 'FlexLift PCスタンド アルミ'),
  ('S-907', 'モバイルバッテリー 10000mAh')
AS t(supplier_item_id, supplier_item_name)
""")

# 列の説明（コメント）を付ける
column_comments = {
    'supplier_item_id'  : '仕入先の商品コード',
    'supplier_item_name': '仕入先の請求書・納品書に書かれた商品名（表記ゆれあり）',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.supplier_items ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

display(spark.table(f"{CATALOG}.{SCHEMA}.supplier_items"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">2つの文章を比べる</div></div>
</div>

```python
df = spark.sql(f"""
SELECT
  ai_similarity('ポータブル電源 PX-500', 'PX500 ポータブル電源（512Wh）')         AS same_product,
  ai_similarity('ポータブル電源 PX-500', 'モバイルバッテリー 10000mAh')            AS similar_category,
  ai_similarity('ポータブル電源 PX-500', '全自動コーヒーメーカー BrewMaster Pro') AS different
""")
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">1つの仕入先の商品と、自社の全商品を比べる</div><div style='color:#555; margin-top:6px; line-height:1.7;'>仕入先の商品（S-901）と、自社の商品マスタの全商品とのスコアを、高い順に並べます。同じ商品がいちばん上に来て、ほかの商品とのスコアとの差が分かります。</div></div>
</div>

```python
supplier_item_id = "S-901"

df = spark.sql(f"""
SELECT
  s.supplier_item_name,
  p.product_name,
  round(ai_similarity(s.supplier_item_name, p.product_name), 3) AS score
FROM {CATALOG}.{SCHEMA}.supplier_items s
CROSS JOIN {CATALOG}.{SCHEMA}.products p
WHERE s.supplier_item_id = :supplier_item_id
ORDER BY score DESC
""", args={"supplier_item_id": supplier_item_id})
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">名寄せ: 仕入先の商品名に一番近い自社商品を探す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>全組み合わせを作って（クロス結合）、仕入先の商品ごとにスコアが最も高い自社商品を選びます。スコアが低いものは「要確認」として人が確認します。</div></div>
</div>

```python
threshold = 0.8   # これ以上なら自動で紐付ける

df = spark.sql(f"""
WITH pairs AS (
  SELECT
    s.supplier_item_id,
    s.supplier_item_name,
    p.product_id,
    p.product_name,
    ai_similarity(s.supplier_item_name, p.product_name) AS score
  FROM {CATALOG}.{SCHEMA}.supplier_items s
  CROSS JOIN {CATALOG}.{SCHEMA}.products p
)
SELECT
  supplier_item_id,
  supplier_item_name,
  max_by(product_id, score)   AS matched_product_id,
  max_by(product_name, score) AS matched_product_name,
  round(max(score), 3)        AS best_score,
  CASE WHEN max(score) >= :threshold THEN '自動で紐付け' ELSE '要確認' END AS judgement
FROM pairs
GROUP BY supplier_item_id, supplier_item_name
ORDER BY supplier_item_id
""", args={"threshold": threshold})
display(df)
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE supplier_item_matching
COMMENT 'supplier_items と products を ai_similarity で突き合わせた名寄せ結果'
AS
WITH pairs AS (
  SELECT s.supplier_item_id, s.supplier_item_name, p.product_id, p.product_name,
         ai_similarity(s.supplier_item_name, p.product_name) AS score
  FROM {CATALOG}.{SCHEMA}.supplier_items s CROSS JOIN {CATALOG}.{SCHEMA}.products p
)
SELECT
  supplier_item_id,
  supplier_item_name,
  max_by(product_id, score)   AS matched_product_id,
  max_by(product_name, score) AS matched_product_name,
  max(score)                  AS best_score
FROM pairs
GROUP BY ALL
""")

# 列の説明（コメント）を付ける
column_comments = {
    'matched_product_id': '最もスコアが高かった自社の商品ID',
    'best_score'        : 'ai_similarity のスコア（0〜1、高いほど似ている）',
}
for col, comment in column_comments.items():
    spark.sql(f"ALTER TABLE {CATALOG}.{SCHEMA}.supplier_item_matching ALTER COLUMN {col} COMMENT :comment", args={"comment": comment})

df = spark.sql(f"""
SELECT * FROM supplier_item_matching ORDER BY best_score DESC
""")
display(df)
```

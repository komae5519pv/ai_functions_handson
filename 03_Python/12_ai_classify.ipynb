```python
from pyspark.sql import functions as F
from datetime import datetime

rows = [
    (1, datetime(2026, 10, 14, 9, 12), 'メール', '注文番号 A-10231 の商品がまだ届きません。10月10日に注文し、到着予定は14日でした。明日のイベントで使うので至急確認してください。山田花子 090-1111-2222'),
    (2, datetime(2026, 10, 14, 10, 40), 'Webフォーム', '請求書の金額が見積と違います。見積では税込132,000円でしたが、届いた請求書は145,200円になっています。確認をお願いします。経理部 佐藤 sato@example.co.jp'),
    (3, datetime(2026, 10, 14, 11, 5), 'チャット', 'ポータブル電源PX-500を使っていたら焦げ臭いにおいがして、本体が異常に熱くなりました。怖くて使えません。どうすればいいですか。'),
    (4, datetime(2026, 10, 14, 13, 22), 'Webフォーム', '会員登録したメールアドレスを変更したいです。新しいアドレスは k.tanaka@example.com です。手順を教えてください。'),
    (5, datetime(2026, 10, 14, 14, 1), 'メール', '先日購入したコーヒーメーカー、とても使いやすく満足しています。店頭スタッフの方の説明も丁寧でした。ありがとうございました。'),
    (6, datetime(2026, 10, 14, 15, 30), 'メール', '法人向けにノートPCを50台まとめて購入したいのですが、ボリュームディスカウントはありますか。来月中に導入したいです。株式会社サンプル 鈴木一郎 03-5555-1234'),
    (7, datetime(2026, 10, 14, 16, 45), 'Webフォーム', '返品したいです。サイズが合いませんでした。未開封です。送り先は 大阪府大阪市北区梅田1-2-3 の高橋です。'),
    (8, datetime(2026, 10, 14, 18, 10), 'チャット', 'アプリにログインできません。パスワードを再設定してもエラーになります。昨日から何度も試していて困っています。'),
]
df = spark.createDataFrame(rows, "inquiry_id INT, received_at TIMESTAMP, channel STRING, body STRING")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'inquiry_id' : '問い合わせID',
    'received_at': '受付日時',
    'channel'    : '受付チャネル（メール / Webフォーム / チャット）',
    'body'       : '問い合わせ本文（お客様が書いた原文。個人情報を含む）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.inquiries")
   .tableProperty("comment", 'カスタマーサポートに届いた問い合わせ（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">ラベル名だけで分類する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>いちばん簡単な形です。ラベルを JSON の配列で渡します。<code>version</code> は <code>2.1</code> を指定するのが推奨です。</div></div>
</div>

```python
from pyspark.sql import functions as F

def options(d):
    # Python の辞書を、AI 関数に渡す map(...) の列に変換する
    return F.create_map(*[F.lit(x) for kv in d.items() for x in kv])

labels = '["配送", "支払い・請求", "製品の故障・安全", "アカウント", "返品・交換", "購入の相談", "その他"]'

inquiries = spark.table(f"{CATALOG}.{SCHEMA}.inquiries")
df = inquiries.select(
    "inquiry_id", "body",
    F.call_function("ai_classify", F.col("body"), F.lit(labels), options({"version": "2.1"})).alias("result"))
display(df)
```

      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">戻り値は VARIANT 型の JSON です。選ばれたラベルは <code>response</code> の配列の中の <code>value</code> に入っています。</div>

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">ラベルに説明を付け、確信度と理由も出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>ラベル名だけでは紛らわしいときは、<b>ラベル: 説明</b> の形で渡すと精度が上がります。<code>enableConfidenceScores</code> と <code>enableRationales</code> を付けると、確信度と理由も返ります。</div></div>
</div>

```python
labels_with_desc = '''{
  "配送":             "商品が届かない、遅れている、お届け日の変更",
  "支払い・請求":     "請求金額、支払い方法、見積との差異",
  "製品の故障・安全": "動かない、壊れた、発熱や異臭など安全に関わること",
  "アカウント":       "ログイン、パスワード、会員情報の変更",
  "返品・交換":       "返品したい、交換したい",
  "購入の相談":       "購入前の質問、まとめ買い、見積の依頼",
  "その他":           "上のどれにも当てはまらないもの（お礼や感想など）"
}'''

df = inquiries.select(
    "inquiry_id", "body",
    F.call_function(
        "ai_classify", F.col("body"), F.lit(labels_with_desc),
        options({"version": "2.1",
                 "instructions": "家電・日用品を扱う EC サイトのサポート窓口に届いた問い合わせです。",
                 "enableConfidenceScores": "true",
                 "enableRationales": "true"})).alias("result"))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">VARIANT から列に取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>F.variant_get(列, "$.response[0].value", "string")</code> のように、JSON のパスと型を指定して取り出します。</div></div>
</div>

```python
classified = inquiries.withColumn(
    "result",
    F.call_function("ai_classify", F.col("body"), F.lit(labels),
                    options({"version": "2.1", "enableConfidenceScores": "true"})))

df = classified.select(
    "inquiry_id", "body",
    F.variant_get("result", "$.response[0].value", "string").alias("category"),
    F.variant_get("result", "$.response[0].confidence_score", "double").alias("confidence"),
    F.variant_get("result", "$.error_message", "string").alias("error_message"))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">6</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">1つの文章に複数のラベルを付ける</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>'multilabel','true'</code> を付けると、当てはまるラベルがすべて返ります。話題が複数ある文章に向いています。</div></div>
</div>

```python
tags = '["急ぎの対応が必要", "お金に関わる", "安全に関わる", "個人情報を含む", "前向きな声"]'

df = (inquiries
      .withColumn("result", F.call_function("ai_classify", F.col("body"), F.lit(tags),
                                            options({"version": "2.1", "multilabel": "true"})))
      # response は配列なので、配列として取り出してから value だけを集める
      .withColumn("tags", F.transform(F.variant_get("result", "$.response", "array<variant>"),
                                      lambda x: F.variant_get(x, "$.value", "string")))
      .select("inquiry_id", "body", "tags"))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">7</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>分類結果は保存しておき、担当部署への振り分けや件数の集計はこのテーブルから行います。</div></div>
</div>

```python
result = inquiries.withColumn(
    "result",
    F.call_function("ai_classify", F.col("body"), F.lit(labels),
                    options({"version": "2.1", "enableConfidenceScores": "true"})))

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'result': 'ai_classify の戻り値（VARIANT）。response[0].value が分類ラベル',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.inquiries_classified")
       .tableProperty("comment", 'inquiries に ai_classify の分類結果を付けたテーブル')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries_classified"))

# 分類ごとの件数
display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries_classified")
        .groupBy(F.variant_get("result", "$.response[0].value", "string").alias("category"))
        .count().orderBy(F.desc("count")))
```

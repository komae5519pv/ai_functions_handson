```python
from pyspark.sql import functions as F
from datetime import datetime

rows = [
    (1, datetime(2026, 10, 14, 9, 12), 'メール', '注文番号 A-10231 の商品がまだ届きません。10月10日に注文し、到着予定は14日でした。明日のイベントで使うので至急確認してください。山田花子 090-1111-2222'),
    (2, datetime(2026, 10, 14, 10, 40), 'Webフォーム', '請求書の金額が見積と違います。見積では税込132,000円でしたが、届いた請求書は145,200円になっています。確認をお願いします。経理部 佐藤 sato@example.co.jp'),
    (3, datetime(2026, 10, 14, 11, 5), 'チャット', 'ポータブル電源PX-500を使っていたら焦げ臭いにおいがして、本体が異常に熱くなりました。怖くて使えません。どうすればいいですか。'),
    (4, datetime(2026, 10, 14, 13, 22), 'Webフォーム', '会員登録したメールアドレスを変更したいです。新しいアドレスは k.tanaka@example.com です。手順を教えてください。'),
    (5, datetime(2026, 10, 14, 14, 1), 'メール', '先日購入したコーヒーメーカー、とても使いやすく満足しています。店頭スタッフの方の説明も丁寧でした。ありがとうございました。'),
    (6, datetime(2026, 10, 14, 15, 30), 'メール', '法人向けにノートPCを50台まとめて購入したいのですが、ボリュームディスカウントはありますか。来月中に導入したいです。株式会社サンプル 鈴木一郎 03-5555-1234'),
    (7, datetime(2026, 10, 14, 16, 45), 'Webフォーム', '返品したいです。サイズが合いませんでした。未開封です。送り先は 大阪府大阪市北区梅田1-2-3 の高橋です。'),
    (8, datetime(2026, 10, 14, 18, 10), 'チャット', 'アプリにログインできません。パスワードを再設定してもエラーになります。昨日から何度も試していて困っています。'),
]
df = spark.createDataFrame(rows, "inquiry_id INT, received_at TIMESTAMP, channel STRING, body STRING")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'inquiry_id' : '問い合わせID',
    'received_at': '受付日時',
    'channel'    : '受付チャネル（メール / Webフォーム / チャット）',
    'body'       : '問い合わせ本文（お客様が書いた原文。個人情報を含む）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.inquiries")
   .tableProperty("comment", 'カスタマーサポートに届いた問い合わせ（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">3つの型の質問を1回で聞く</div><div style='color:#555; margin-top:6px; line-height:1.7;'>「至急対応か（noul）」「どのチームが担当か（choice）」「緊急度（score）」を同時に聞きます。</div></div>

```python
from pyspark.sql import functions as F

questions_full = '''{
  "needs_escalation": {
    "type": "noul",
    "instructions": "この問い合わせは、すぐに上長や専門部署へ引き継ぐべきですか",
    "criteria": {
      "true":  "安全に関わる、お客様の業務やイベントに支障が出る、金銭トラブルになりそう",
      "false": "通常の手順で対応でき、急ぎではない"
    }
  },
  "team": {
    "type": "choice",
    "instructions": "どのチームが担当すべきですか",
    "criteria": {
      "logistics": "配送・お届け日に関すること",
      "billing":   "請求・支払い・見積に関すること",
      "product":   "製品の故障・安全・使い方",
      "account":   "ログイン・会員情報",
      "sales":     "購入前の相談・法人のまとめ買い",
      "none":      "対応不要（お礼や感想）"
    }
  },
  "urgency": {
    "type": "score",
    "instructions": "どのくらい急いで対応すべきですか",
    "criteria": ["急ぎではない（1週間以内でよい）", "数日以内に対応", "今日中に対応", "すぐに対応（安全・損害に関わる）"]
  }
}'''

inquiries = spark.table(f"{CATALOG}.{SCHEMA}.inquiries")
df = inquiries.select("inquiry_id", "body",
                      F.call_function("ai_decide", F.col("body"), F.lit(questions_full)).alias("decision"))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">答えを列に取り出し、しきい値で振り分ける</div><div style='color:#555; margin-top:6px; line-height:1.7;'>確率やスコアは数値なので、条件分岐にそのまま使えます。</div></div>
</div>

```python
questions = '''{
  "needs_escalation": {"type": "noul", "instructions": "すぐに上長や専門部署へ引き継ぐべきですか"},
  "team": {"type": "choice", "instructions": "担当チーム",
           "criteria": {"logistics": "配送", "billing": "請求・支払い", "product": "製品の故障・安全", "account": "ログイン・会員情報", "sales": "購入前の相談", "none": "対応不要"}},
  "urgency": {"type": "score", "instructions": "対応の緊急度", "criteria": ["急ぎではない", "数日以内", "今日中", "すぐに"]}
}'''
escalation_threshold = 0.7   # この確率以上ならエスカレーション
today_threshold = 2.0        # 緊急度スコアがこれ以上なら今日中

decided = inquiries.withColumn("d", F.call_function("ai_decide", F.col("body"), F.lit(questions)))
answer = lambda path, typ: F.variant_get("d", f"$.response.answers.{path}", typ)

df = (decided.select(
        "inquiry_id", "body",
        F.round(answer("needs_escalation.probability", "double"), 2).alias("escalation_prob"),
        answer("team.choice", "string").alias("team"),
        F.round(answer("team.confidence", "double"), 2).alias("team_confidence"),
        F.round(answer("urgency.score", "double"), 2).alias("urgency_score"))
      .withColumn("routing",
                  F.when(F.col("escalation_prob") >= escalation_threshold, "🚨 エスカレーション")
                   .when(F.col("urgency_score") >= today_threshold, "⏱ 今日中")
                   .otherwise("通常"))
      .orderBy(F.desc("urgency_score")))
display(df)
```

```python
result = inquiries.withColumn("decision", F.call_function("ai_decide", F.col("body"), F.lit(questions)))

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'decision': 'ai_decide の戻り値（VARIANT）。response.answers に質問ごとの答え',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.inquiries_decided")
       .tableProperty("comment", 'inquiries に ai_decide でエスカレーション確率・担当チーム・緊急度を付けたテーブル')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries_decided"))
```

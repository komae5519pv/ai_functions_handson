```python
from pyspark.sql import functions as F
from datetime import datetime

rows = [
    (1, datetime(2026, 10, 14, 9, 12), 'メール', '注文番号 A-10231 の商品がまだ届きません。10月10日に注文し、到着予定は14日でした。明日のイベントで使うので至急確認してください。山田花子 090-1111-2222'),
    (2, datetime(2026, 10, 14, 10, 40), 'Webフォーム', '請求書の金額が見積と違います。見積では税込132,000円でしたが、届いた請求書は145,200円になっています。確認をお願いします。経理部 佐藤 sato@example.co.jp'),
    (3, datetime(2026, 10, 14, 11, 5), 'チャット', 'ポータブル電源PX-500を使っていたら焦げ臭いにおいがして、本体が異常に熱くなりました。怖くて使えません。どうすればいいですか。'),
    (4, datetime(2026, 10, 14, 13, 22), 'Webフォーム', '会員登録したメールアドレスを変更したいです。新しいアドレスは k.tanaka@example.com です。手順を教えてください。'),
    (5, datetime(2026, 10, 14, 14, 1), 'メール', '先日購入したコーヒーメーカー、とても使いやすく満足しています。店頭スタッフの方の説明も丁寧でした。ありがとうございました。'),
    (6, datetime(2026, 10, 14, 15, 30), 'メール', '法人向けにノートPCを50台まとめて購入したいのですが、ボリュームディスカウントはありますか。来月中に導入したいです。株式会社サンプル 鈴木一郎 03-5555-1234'),
    (7, datetime(2026, 10, 14, 16, 45), 'Webフォーム', '返品したいです。サイズが合いませんでした。未開封です。送り先は 大阪府大阪市北区梅田1-2-3 の高橋です。'),
    (8, datetime(2026, 10, 14, 18, 10), 'チャット', 'アプリにログインできません。パスワードを再設定してもエラーになります。昨日から何度も試していて困っています。'),
]
df = spark.createDataFrame(rows, "inquiry_id INT, received_at TIMESTAMP, channel STRING, body STRING")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'inquiry_id' : '問い合わせID',
    'received_at': '受付日時',
    'channel'    : '受付チャネル（メール / Webフォーム / チャット）',
    'body'       : '問い合わせ本文（お客様が書いた原文。個人情報を含む）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.inquiries")
   .tableProperty("comment", 'カスタマーサポートに届いた問い合わせ（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">項目名だけで取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>いちばん簡単な形です。欲しい項目の名前を配列で渡します。見つからない項目は <code>null</code> になります。</div></div>

```python
from pyspark.sql import functions as F

def options(d):
    # Python の辞書を、AI 関数に渡す map(...) の列に変換する
    return F.create_map(*[F.lit(x) for kv in d.items() for x in kv])

schema = '["注文番号", "製品名", "氏名", "電話番号", "メールアドレス"]'

inquiries = spark.table(f"{CATALOG}.{SCHEMA}.inquiries")
df = inquiries.select(
    "inquiry_id", "body",
    F.call_function("ai_extract", F.col("body"), F.lit(schema), options({"version": "2.1"})).alias("result"))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">型と説明を付けて取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>項目ごとに <code>type</code> と <code>description</code> を付けると、数値は数値として、選択肢は決めた値の中から返ります。<code>enum</code> を使うと値を選択肢に限定できます。</div></div>
</div>

```python
typed_schema = '''{
  "order_number":   {"type": "string",  "description": "注文番号（例: A-10231）。なければ null"},
  "amount_jpy":     {"type": "integer", "description": "文中に出てくる金額のうち、お客様が主張している金額（円）"},
  "quantity":       {"type": "integer", "description": "購入・注文したい数量"},
  "needs_reply_by": {"type": "string",  "description": "回答や対応の期限。文中に日付や「明日」などがあれば書く"},
  "customer_type":  {"type": "enum", "labels": ["個人", "法人"], "description": "問い合わせてきたのが個人か法人か"},
  "contact": {
    "type": "object",
    "properties": {
      "name":  {"type": "string", "description": "氏名"},
      "phone": {"type": "string", "description": "電話番号"},
      "email": {"type": "string", "description": "メールアドレス"}
    }
  }
}'''

df = inquiries.select(
    "inquiry_id", "body",
    F.call_function(
        "ai_extract", F.col("body"), F.lit(typed_schema),
        options({"version": "2.1",
                 "instructions": "EC サイトのサポート窓口に届いた問い合わせです。書かれていない項目は推測せず null にしてください。"})).alias("result"))
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">確信度と引用を付ける</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>enableConfidenceScores</code> で項目ごとの確信度（0〜1）、<code>enableCitations</code> で原文のどの範囲から取ったか（<code>metadata.citations</code> の start / stop は文字位置）が返ります。人が確認すべき行を絞り込むのに使えます。</div></div>
</div>

```python
small_schema = '{"order_number": {"type": "string"}, "amount_jpy": {"type": "integer"}}'

df = (inquiries.where(F.col("inquiry_id").isin(1, 2))
      .select("inquiry_id", "body",
              F.call_function("ai_extract", F.col("body"), F.lit(small_schema),
                              options({"version": "2.1", "enableConfidenceScores": "true", "enableCitations": "true"})).alias("result")))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">6</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">VARIANT から列に取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>F.variant_get(列, "$.response.項目.value", "型")</code> で列にします。入れ子の項目は <code>$.response.contact.name.value</code> のようにたどります。</div></div>
</div>

```python
save_schema = '''{
  "order_number": {"type": "string"},
  "amount_jpy":   {"type": "integer"},
  "contact": {"type": "object", "properties": {"name": {"type": "string"}, "phone": {"type": "string"}, "email": {"type": "string"}}}
}'''

# 取り出したい項目と、その JSON パス・型
fields = {
    "order_number":  ("$.response.order_number.value",  "string"),
    "amount_jpy":    ("$.response.amount_jpy.value",    "int"),
    "contact_name":  ("$.response.contact.name.value",  "string"),
    "contact_phone": ("$.response.contact.phone.value", "string"),
    "contact_email": ("$.response.contact.email.value", "string"),
}

extracted = inquiries.withColumn(
    "result", F.call_function("ai_extract", F.col("body"), F.lit(save_schema), options({"version": "2.1"})))
df = extracted.select("inquiry_id", *[F.variant_get("result", path, typ).alias(name) for name, (path, typ) in fields.items()])
display(df.orderBy("inquiry_id"))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">7</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div></div>
</div>

```python
result = (inquiries
          .withColumn("result", F.call_function("ai_extract", F.col("body"), F.lit(save_schema), options({"version": "2.1"})))
          .select("inquiry_id", "received_at", "channel", "body",
                  *[F.variant_get("result", path, typ).alias(name) for name, (path, typ) in fields.items()]))

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'order_number': 'AI が本文から取り出した注文番号',
    'amount_jpy': 'AI が本文から取り出した金額（円）',
    'contact_name': 'AI が本文から取り出した氏名',
    'contact_phone': 'AI が本文から取り出した電話番号',
    'contact_email': 'AI が本文から取り出したメールアドレス',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.inquiries_extracted")
       .tableProperty("comment", 'inquiries から ai_extract で注文番号・金額・連絡先を取り出したテーブル')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries_extracted"))
```

```python
from pyspark.sql import functions as F

policy_markdown = """# 返品・交換ポリシー
## 1. 返品できる期間
商品到着後14日以内で、未開封のものに限ります。
## 2. 返品できないもの
開封済みの消耗品、お客様の責任で傷がついたもの、セール品。
## 3. 送料
お客様都合の返品はお客様負担、初期不良の場合は当社が負担します。"""

df = spark.createDataFrame([(policy_markdown,)], "text STRING").select(
    F.call_function("ai_prep_search", F.col("text")).alias("prepped"))
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">PDF を読み取ってから分割する</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>ai_parse_document</code> の結果を渡すと、見出し・表・ページ番号を踏まえて分割され、表には要約と「この表で答えられる質問」が付きます。</div></div>
</div>

```python
docs = spark.read.format("binaryFile").load(f"{VOL_ROOT}/docs/")
prepped = docs.select(
    "path",
    F.call_function("ai_prep_search", F.call_function("ai_parse_document", F.col("content"))).alias("prepped"))
display(prepped)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">チャンクを1行ずつに展開し、メタデータも付ける</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>schema</code> オプションで、文書から取り出した値（製品名・文書の種類など）を全チャンクに付けられます。AI Search で絞り込み検索に使えます。</div></div>
</div>

```python
def options(d):
    # Python の辞書を、AI 関数に渡す map(...) の列に変換する
    return F.create_map(*[F.lit(x) for kv in d.items() for x in kv])

doc_metadata_schema = '{"product_name": {"type": "string", "description": "対象の製品名"}, "document_type": {"type": "string", "description": "文書の種類"}}'

prepped_meta = docs.select(
    "path",
    F.call_function("ai_prep_search",
                    F.call_function("ai_parse_document", F.col("content"), options({"version": "2.0"})),
                    options({"schema": doc_metadata_schema})).alias("result"))

chunks = (prepped_meta
          .select("path", F.explode(F.variant_get("result", "$.document.contents", "array<variant>")).alias("c"))
          .select(F.variant_get("c", "$.chunk_id", "string").alias("chunk_id"),
                  F.variant_get("c", "$.chunk_position", "int").alias("chunk_position"),
                  F.variant_get("c", "$.metadata.product_name", "string").alias("product_name"),
                  F.variant_get("c", "$.metadata.document_type", "string").alias("document_type"),
                  F.variant_get("c", "$.chunk_to_retrieve", "string").alias("chunk_to_retrieve"),
                  F.variant_get("c", "$.chunk_to_embed", "string").alias("chunk_to_embed"),
                  F.col("path").alias("source_uri")))
display(chunks.orderBy("chunk_position"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">AI Search のソース用テーブルとして保存する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>このテーブルをもとに AI Search のインデックスを作ります（<code>41_vector_search</code>）。インデックスの同期には変更データフィード（Change Data Feed）を有効にしておきます。</div></div>

```python
result = chunks.drop("document_type")

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'chunk_id': 'チャンクのID（インデックスの主キー）',
    'chunk_to_retrieve': 'チャンクの本文（検索結果として LLM に渡す）',
    'chunk_to_embed': '文脈を足した本文（ベクトル化に使う）',
    'source_uri': '元の文書のパス',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.document_chunks")
       .tableProperty("delta.enableChangeDataFeed", "true")
       .tableProperty("comment", '製品マニュアルを ai_prep_search で分割したチャンク（AI Search のソース用）')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.document_chunks"))
```

```python
from pyspark.sql import functions as F

rows = [
    ('P001', 'ポータブル電源 PX-500', '家電', '容量512Wh / AC出力500W / 重量6.2kg / ソーラー充電対応', 69800),
    ('P002', 'ワイヤレスイヤホン AirTone S2', 'オーディオ', 'ノイズキャンセリング / 最大30時間再生 / 防水IPX4', 14800),
    ('P003', '電動アシスト自転車 RideOn 26', 'モビリティ', '26インチ / 航続距離80km / 7段変速', 128000),
    ('P004', '全自動コーヒーメーカー BrewMaster Pro', 'キッチン家電', 'ミル付き / 最大10杯 / タイマー予約', 32800),
    ('P005', 'ノートPCスタンド FlexLift', 'オフィス用品', 'アルミ製 / 高さ6段階 / 折りたたみ式', 4980),
    ('P006', 'ロボット掃除機 CleanBot X', '生活家電', 'マッピング機能 / 自動ゴミ収集 / 水拭き対応', 59800),
]
df = spark.createDataFrame(rows, "product_id STRING, product_name STRING, category STRING, spec STRING, price_jpy INT")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'product_id'  : '商品ID',
    'product_name': '自社の正式な商品名',
    'category'    : '商品カテゴリ',
    'spec'        : '主な仕様（社内の商品マスタに登録された短いメモ）',
    'price_jpy'   : '税込販売価格（円）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.products")
   .tableProperty("comment", '自社の商品マスタ（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.products"))
```

```python
from pyspark.sql import functions as F

rows = [
    ('S-901', 'PX500 ポータブル電源（512Wh）'),
    ('S-902', 'AirTone S2 ワイヤレス イヤホン 黒'),
    ('S-903', 'コーヒーメーカー BrewMaster（ミル付）'),
    ('S-904', 'CleanBot-X ロボットクリーナー'),
    ('S-905', '電動自転車 ライドオン 26インチ'),
    ('S-906', 'FlexLift PCスタンド アルミ'),
    ('S-907', 'モバイルバッテリー 10000mAh'),
]
df = spark.createDataFrame(rows, "supplier_item_id STRING, supplier_item_name STRING")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'supplier_item_id'  : '仕入先の商品コード',
    'supplier_item_name': '仕入先の請求書・納品書に書かれた商品名（表記ゆれあり）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.supplier_items")
   .tableProperty("comment", '仕入先から届いた納品データ。自社の商品マスタとは表記が違う（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.supplier_items"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">2つの文章を比べる</div></div>
</div>

```python
from pyspark.sql import functions as F

base = "ポータブル電源 PX-500"
candidates = ["PX500 ポータブル電源（512Wh）", "モバイルバッテリー 10000mAh", "全自動コーヒーメーカー BrewMaster Pro"]
df = (spark.createDataFrame([(base, c) for c in candidates], "text1 STRING, text2 STRING")
      .withColumn("score", F.call_function("ai_similarity", F.col("text1"), F.col("text2"))))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">1つの仕入先の商品と、自社の全商品を比べる</div><div style='color:#555; margin-top:6px; line-height:1.7;'>仕入先の商品（S-901）と、自社の商品マスタの全商品とのスコアを、高い順に並べます。同じ商品がいちばん上に来て、ほかの商品とのスコアとの差が分かります。</div></div>
</div>

```python
supplier_item_id = "S-901"

products = spark.table(f"{CATALOG}.{SCHEMA}.products")
supplier_items = spark.table(f"{CATALOG}.{SCHEMA}.supplier_items").filter(F.col("supplier_item_id") == supplier_item_id)

df = (supplier_items.crossJoin(products)
      .select("supplier_item_name", "product_name",
              F.round(F.call_function("ai_similarity", F.col("supplier_item_name"), F.col("product_name")), 3).alias("score"))
      .orderBy(F.col("score").desc()))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">名寄せ: 仕入先の商品名に一番近い自社商品を探す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>全組み合わせを作って（クロス結合）、仕入先の商品ごとにスコアが最も高い自社商品を選びます。スコアが低いものは「要確認」として人が確認します。</div></div>
</div>

```python
threshold = 0.8   # これ以上なら自動で紐付ける

products = spark.table(f"{CATALOG}.{SCHEMA}.products")
supplier_items = spark.table(f"{CATALOG}.{SCHEMA}.supplier_items")

pairs = (supplier_items.crossJoin(products)
         .withColumn("score", F.call_function("ai_similarity", F.col("supplier_item_name"), F.col("product_name"))))
matched = (pairs.groupBy("supplier_item_id", "supplier_item_name")
           .agg(F.max_by("product_id", "score").alias("matched_product_id"),
                F.max_by("product_name", "score").alias("matched_product_name"),
                F.round(F.max("score"), 3).alias("best_score"))
           .withColumn("judgement", F.when(F.col("best_score") >= threshold, "自動で紐付け").otherwise("要確認")))
display(matched.orderBy("supplier_item_id"))
```

    <div>
      <strong style="color: #e65100; font-size: 1.08em;">言語や文字種をそろえてから比べる</strong>
      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">日本語と英語（例: 「プリウス」と「PRIUS」）のように<b>言語や文字種が違うと、同じものでもスコアが低く出る</b>ことがあります（作成時の実測で 0.50）。比べる前に表記をそろえる（片方を翻訳する、カナに統一する）と安定します。</div>

```python
result = matched.drop("judgement")

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'matched_product_id': '最もスコアが高かった自社の商品ID',
    'best_score': 'ai_similarity のスコア（0〜1、高いほど似ている）',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.supplier_item_matching")
       .tableProperty("comment", 'supplier_items と products を ai_similarity で突き合わせた名寄せ結果')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.supplier_item_matching"))
```

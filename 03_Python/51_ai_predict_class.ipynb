  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">2</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>サブスク会員 400 人の利用状況と解約実績を作ります。最後の 20 人は <code>churned</code> が NULL（まだ結果が出ていない）で、ここを予測します。</div></div>
</div>

```python
from pyspark.sql import functions as F

def h(salt, mod):
    return F.abs(F.hash(F.col("id"), F.lit(salt))) % mod

f = spark.range(400).select(
    (F.col("id") + 1).alias("member_id"),
    F.when(h("p", 3) == 0, "年額").otherwise("月額").alias("plan"),
    (1 + h("t", 48)).cast("int").alias("tenure_months"),
    h("l", 30).cast("int").alias("monthly_logins"),
    h("s", 6).cast("int").alias("support_tickets"),
    h("b", 180).cast("int").alias("days_since_last_purchase"),
    (h("n", 100) / 100.0).alias("noise"))

risk = (F.when(F.col("plan") == "月額", 1.5).otherwise(0.0)
        + F.when(F.col("tenure_months") < 6, 1.5).otherwise(0.0)
        + F.when(F.col("monthly_logins") < 5, 2.0).otherwise(0.0)
        + F.col("support_tickets") * 0.5
        + F.col("days_since_last_purchase") / 60.0
        + F.col("noise"))
df = (f.withColumn("churned", F.when(F.col("member_id") > 380, F.lit(None))   # 最後の20人は結果未確定（＝予測対象）
                               .otherwise(risk > 4.5))
       .drop("noise"))

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'member_id'               : '会員ID',
    'plan'                    : '契約プラン（月額 / 年額）',
    'tenure_months'           : '契約してからの月数',
    'monthly_logins'          : '直近1か月のログイン回数',
    'support_tickets'         : '直近3か月の問い合わせ件数',
    'days_since_last_purchase': '最後に購入してからの日数',
    'churned'                 : '解約したか（true / false）。NULL はまだ結果が出ていない＝予測したい会員',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.members")
   .tableProperty("comment", 'サブスク会員の利用状況と解約実績（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.members"))
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">解約を予測する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>答えの列と手がかりの列を指定するだけです。答えが NULL の 20 人について、予測と確率が返ります。</div></div>
</div>

```python
from pyspark.sql import functions as F

target_col = "churned"
feature_cols = ["plan", "tenure_months", "monthly_logins", "support_tickets", "days_since_last_purchase"]
seed = 42

members = spark.table(f"{CATALOG}.{SCHEMA}.members")

def predict_class(src_df):
    # テーブル値関数なので spark.sql に DataFrame を渡して呼ぶ
    features = ", ".join(f"'{c}'" for c in feature_cols)
    return spark.sql(
        f"SELECT * FROM ai_predict_class(input => TABLE({{src}}), target_col => '{target_col}', "
        f"feature_cols => ARRAY({features}), seed => {seed})",
        src=src_df)

predicted = predict_class(members)
display(predicted.orderBy(F.desc("churned_confidence")))
```

  <div style="display: flex; align-items: flex-start; gap: 12px;">
    <span style="font-size: 22px;">👀</span>
    <div>
      <strong style="color: #00695c; font-size: 1.08em;">ここを見る</strong>
      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">返ってくるのは答えが NULL だった 20 行だけです。<code>churned_confidence</code> は「予測した分類である確率」なので、確率の高い順に引き止め施策の対象にする、といった使い方ができます。</div>
    </div>

</div>

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">どのくらい当たるかを確かめる</div><div style='color:#555; margin-top:6px; line-height:1.7;'>答えがわかっている行の一部をわざと NULL にして予測させ、本当の答えと比べます（ホールドアウト検証）。ここでは会員 ID が 5 の倍数の人を検証用にします。</div></div>
</div>

```python
holdout = (members.where(F.col("churned").isNotNull())
           .withColumn("actual", F.col("churned"))
           .withColumn("churned", F.when(F.col("member_id") % 5 == 0, F.lit(None)).otherwise(F.col("churned"))))

checked = predict_class(holdout)
df = checked.agg(F.count("*").alias("test_rows"),
                 F.round(F.avg((F.col("churned_prediction") == F.col("actual")).cast("double")), 3).alias("accuracy"))
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">予測結果をテーブルに保存する</div></div>
</div>

```python
result = predicted.select("member_id", "churned_prediction", "churned_confidence")

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'churned_prediction': '予測した解約の有無',
    'churned_confidence': '予測した分類である確率（0〜1）',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.members_churn_prediction")
       .tableProperty("comment", 'members の未確定の会員について ai_predict_class で解約を予測した結果')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.members_churn_prediction"))
```

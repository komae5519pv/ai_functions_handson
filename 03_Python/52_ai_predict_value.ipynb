  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">2</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>既存店 140 店の条件と月商、出店候補地 10 か所の条件を作ります。候補地は <code>monthly_sales</code> が NULL です。</div></div>

```python
from pyspark.sql import functions as F

def h(salt, mod, base=F.col("id")):
    return F.abs(F.hash(base, F.lit(salt))) % mod

f = spark.range(150).select(
    (F.col("id") + 1).alias("location_id"),
    F.element_at(F.array(F.lit("駅前"), F.lit("郊外"), F.lit("商業施設内")), h("a", 3).cast("int") + 1).alias("area_type"),
    (80 + h("f", 420)).cast("int").alias("floor_area_m2"),
    (1 + h("w", 25)).cast("int").alias("walk_min_from_station"),
    h("p", 120).cast("int").alias("parking_spaces"),
    (3000 + h("o", 45000)).cast("int").alias("population_1km"),
    (0.9 + h("n", 200, F.col("id") + 1) / 1000.0).alias("noise"))

area_bonus = (F.when(F.col("area_type") == "駅前", 4000000)
               .when(F.col("area_type") == "商業施設内", 2500000).otherwise(0))
sales = (F.col("floor_area_m2") * 38000 + F.col("population_1km") * 120
         - F.col("walk_min_from_station") * 250000 + F.col("parking_spaces") * 30000 + area_bonus) * F.col("noise")
df = (f.withColumn("monthly_sales", F.when(F.col("location_id") > 140, F.lit(None))   # 最後の10か所は出店前（＝予測対象）
                                     .otherwise(F.round(sales, -4).cast("bigint")))
       .drop("noise"))

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'location_id'          : '店舗・候補地ID',
    'area_type'            : '立地タイプ（駅前 / 郊外 / 商業施設内）',
    'floor_area_m2'        : '売場面積（平方メートル）',
    'walk_min_from_station': '最寄り駅からの徒歩分数',
    'parking_spaces'       : '駐車場の台数',
    'population_1km'       : '半径1km の人口（人）',
    'monthly_sales'        : '月商（円）。NULL はまだ出店していない候補地＝予測したい行',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.store_locations")
   .tableProperty("comment", '既存店の実績と出店候補地の条件（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.store_locations"))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">候補地の月商を予測する</div></div>
</div>

```python
from pyspark.sql import functions as F

target_col = "monthly_sales"
feature_cols = ["area_type", "floor_area_m2", "walk_min_from_station", "parking_spaces", "population_1km"]
seed = 42

locations = spark.table(f"{CATALOG}.{SCHEMA}.store_locations")

def predict_value(src_df, target=target_col):
    # テーブル値関数なので spark.sql に DataFrame を渡して呼ぶ
    features = ", ".join(f"'{c}'" for c in feature_cols)
    return spark.sql(
        f"SELECT * FROM ai_predict_value(input => TABLE({{src}}), target_col => '{target}', "
        f"feature_cols => ARRAY({features}), seed => {seed})",
        src=src_df)

predicted = predict_value(locations)
display(predicted.withColumn("monthly_sales_prediction", F.round("monthly_sales_prediction", -4))
                 .orderBy(F.desc("monthly_sales_prediction")))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">どのくらい当たるかを確かめる</div><div style='color:#555; margin-top:6px; line-height:1.7;'>既存店の一部（ID が 5 の倍数）の月商をわざと隠して予測させ、実際の月商との誤差（平均絶対誤差・平均誤差率）を出します。</div></div>
</div>

```python
holdout = (locations.where(F.col("monthly_sales").isNotNull())
           .withColumn("actual", F.col("monthly_sales"))
           .withColumn("monthly_sales", F.when(F.col("location_id") % 5 == 0, F.lit(None)).otherwise(F.col("monthly_sales"))))

checked = predict_value(holdout)
err = F.abs(F.col("monthly_sales_prediction") - F.col("actual"))
df = checked.agg(F.count("*").alias("test_rows"),
                 F.round(F.avg(err), -3).alias("mean_abs_error_jpy"),
                 F.round(F.avg(err / F.col("actual")), 3).alias("mean_abs_pct_error"))
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">予測結果をテーブルに保存する</div></div>
</div>

```python
result = predicted.select("location_id", "area_type", "monthly_sales_prediction")

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'monthly_sales_prediction': '予測した月商（円）',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.store_sales_prediction")
       .tableProperty("comment", 'store_locations の出店候補地について ai_predict_value で月商を予測した結果')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.store_sales_prediction"))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">6</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">マーケティングの例: 会員ごとに、この先3か月の購入額を予測する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>会員ごとの「この先いくら買ってくれそうか」がわかると、特典の厚さや、営業・広告の力のかけ方を会員ごとに変えられます。過去1年の購入回数・平均購入額・最後に買ってからの日数・メールの開封率・アプリを使っているかを手がかりに、その後3か月の購入額を予測します。</div></div>
</div>
<div style="border-left: 4px solid #7b1fa2; background: #f3e5f5; padding: 14px 20px; border-radius: 4px; margin: 14px 0;">
  <div style="display: flex; align-items: flex-start; gap: 12px;">
    <span style="font-size: 22px;">📝</span>
    <div>
      <strong style="color: #4a148c; font-size: 1.08em;">サンプルデータ</strong>
      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">会員 1,000 人の購買の様子と、その後 3 か月の購入額です。最後の 50 人（<code>member_id</code> 951〜1000）は入会したばかりで <code>next_90d_amount</code> が NULL です。ここを予測します。</div>
    </div>
  </div>
</div>

```python
from pyspark.sql import functions as F

def h(salt, mod):
    return F.abs(F.hash(F.col("id"), F.lit(salt))) % mod

f = spark.range(1000).select(
    (F.col("id") + 1).alias("member_id"),
    (1 + h("o", 24)).cast("int").alias("orders_12m"),
    (2000 + h("a", 13000)).cast("int").alias("avg_order_jpy"),
    (1 + h("d", 365)).cast("int").alias("days_since_last"),
    F.round(h("e", 600) / 1000.0, 2).alias("email_open_rate"),
    (h("p", 2) == 0).alias("app_user"),
    (0.85 + h("n", 300) / 1000.0).alias("noise"))

recency = (F.when(F.col("days_since_last") <= 90, 1.0)
            .when(F.col("days_since_last") <= 180, 0.6).otherwise(0.3))
amount = (F.col("orders_12m") / 4.0 * F.col("avg_order_jpy") * recency * (0.8 + F.col("email_open_rate"))
          * F.when(F.col("app_user"), 1.2).otherwise(1.0) * F.col("noise"))
df = (f.withColumn("next_90d_amount", F.when(F.col("member_id") > 950, F.lit(None))   # 最後の50人はまだ結果が出ていない
                                       .otherwise(F.round(amount, -2).cast("bigint")))
       .drop("noise"))

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'member_id'      : '会員ID',
    'orders_12m'     : '過去1年の購入回数',
    'avg_order_jpy'  : '1回あたりの平均購入額（円）',
    'days_since_last': '最後に買ってからの日数',
    'email_open_rate': 'メールの開封率（0〜1）',
    'app_user'       : 'アプリを使っているか',
    'next_90d_amount': 'その後3か月の購入額（円）。NULL はまだ結果が出ていない会員＝予測したい行',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.member_purchase")
   .tableProperty("comment", '会員ごとの購買の様子と、その後3か月の購入額（AI Functions デモ用の架空データ。1,000人。最後の50人は入会したばかりで、まだ結果が出ていない）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.member_purchase"))
```

```python
members = spark.table(f"{CATALOG}.{SCHEMA}.member_purchase")
member_features = ["orders_12m", "avg_order_jpy", "days_since_last", "email_open_rate", "app_user"]

def predict_amount(src_df, target):
    features = ", ".join(f"'{c}'" for c in member_features)
    return spark.sql(
        f"SELECT * FROM ai_predict_value(input => TABLE({{src}}), target_col => '{target}', "
        f"feature_cols => ARRAY({features}), seed => {seed})",
        src=src_df)

display(predict_amount(members, "next_90d_amount")
        .withColumn("next_90d_amount_prediction", F.round("next_90d_amount_prediction", -2))
        .orderBy(F.desc("next_90d_amount_prediction")))
```

```python
known = (members.where(F.col("next_90d_amount").isNotNull())
         .withColumn("actual", F.col("next_90d_amount"))
         .withColumn("amount_masked", F.when(F.col("member_id") % 5 == 0, F.lit(None)).otherwise(F.col("next_90d_amount")))
         .drop("next_90d_amount"))

checked = predict_amount(known, "amount_masked")
err = F.abs(F.col("amount_masked_prediction") - F.col("actual"))
display(checked.agg(F.count("*").alias("test_members"),
                    F.round(F.avg("actual")).alias("avg_actual_jpy"),
                    F.round(F.avg(err)).alias("mean_abs_error_jpy"),
                    F.round(F.sum(err) / F.sum("actual"), 3).alias("weighted_abs_pct_error")))
```

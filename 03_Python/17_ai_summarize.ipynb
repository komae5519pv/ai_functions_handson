  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>営業担当が書いた顧客訪問メモ 3 件を作ります。</div></div>

```python
from pyspark.sql import functions as F

rows = [
    (1, '株式会社サンプル', '本日、総務部の鈴木様と面談。来月の新入社員研修に合わせてノートPCを50台導入したいとのこと。現在は他社製品を使っているが、サポートの対応が遅く不満がある。価格よりも、故障時にすぐ代替機を出せるかを重視している。予算は1台あたり15万円以内。見積は今週中に欲しいと言われた。次回は情報システム部の方も同席予定。'),
    (2, 'みなと商店', '店主の田中様より、店舗の照明をLEDに切り替えたいと相談。電気代の削減が目的。営業時間中の工事は避けたいので、定休日の水曜に作業したい。照明は全部で40本程度。補助金が使えるかどうかも知りたいとのこと。前回提案した防犯カメラは今回は見送り。'),
    (3, '北斗フーズ', '物流センターの佐藤様と打ち合わせ。冷凍倉庫の温度管理を自動化したい。今は1日3回、手書きで記録している。温度が上がったときにすぐ通知が欲しい。センサーは10か所ほど。既存のWi-Fiが届かないエリアがあるので、通信方式の確認が必要。年内に試験導入して、効果があれば他の2拠点にも広げたい。'),
]
df = spark.createDataFrame(rows, "note_id INT, customer STRING, note STRING")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'note_id' : 'メモID',
    'customer': '訪問先の顧客名（架空）',
    'note'    : '営業担当が書いた訪問メモの本文',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.visit_notes")
   .tableProperty("comment", '営業担当が書いた顧客訪問メモ（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.visit_notes"))
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">既定の長さで要約する</div></div>
</div>

```python
from pyspark.sql import functions as F

notes = spark.table(f"{CATALOG}.{SCHEMA}.visit_notes")
df = notes.withColumn("summary", F.call_function("ai_summarize", F.col("note")))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">長さを変えて比べる</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>max_words</code> で長さの目安を変えられます。</div></div>
</div>

```python
short_words, long_words = 10, 40   # 要約の長さの目安（語数）

df = notes.select(
    "note_id", "customer",
    F.call_function("ai_summarize", F.col("note"), F.lit(short_words)).alias("short_summary"),
    F.call_function("ai_summarize", F.col("note"), F.lit(long_words)).alias("long_summary"))
display(df)
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">日本語で確実に要約する（ai_query）</div><div style='color:#555; margin-top:6px; line-height:1.7;'>モデルを選べる <code>ai_query</code> なら、言語・長さ・観点を指示文で決められます。</div></div>
</div>

```python
prompt = F.concat(F.lit("次の営業メモを、日本語で60文字以内に要約してください。要約文だけを出力してください。\n\n"), F.col("note"))
df = notes.select("note_id", "customer", F.call_function("ai_query", F.lit(LLM_ENDPOINT), prompt).alias("summary_ja"))
display(df)
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div></div>

```python
result = notes.withColumn("summary", F.call_function("ai_summarize", F.col("note"), F.lit(30)))

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'summary': 'AI が要約した訪問メモ（30語程度）',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.visit_notes_summary")
       .tableProperty("comment", 'visit_notes を ai_summarize で要約したテーブル')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.visit_notes_summary"))
```

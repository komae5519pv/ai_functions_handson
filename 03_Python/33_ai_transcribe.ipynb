```python
from pyspark.sql import functions as F

audio = spark.read.format("binaryFile").load(f"{VOL_ROOT}/audio/")
transcribed = audio.select(
    F.regexp_extract("path", r"[^/]+$", 0).alias("file_name"),
    F.call_function("ai_transcribe", F.col("content")).alias("transcription"))
display(transcribed)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">発話ごとに1行にする</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>segments</code> を行に展開すると、誰がいつ何を話したかの表になります。</div></div>
</div>

```python
segments = (transcribed
            .select("file_name", F.explode(F.variant_get("transcription", "$.response.segments", "array<variant>")).alias("seg"))
            .select("file_name",
                    F.round(F.variant_get("seg", "$.start", "double"), 1).alias("start_sec"),
                    F.round(F.variant_get("seg", "$.end", "double"), 1).alias("end_sec"),
                    F.variant_get("seg", "$.speaker_id", "string").alias("speaker"),
                    F.variant_get("seg", "$.text", "string").alias("text")))
display(segments.orderBy("file_name", "start_sec"))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">全文にして、ほかの AI 関数につなげる</div><div style='color:#555; margin-top:6px; line-height:1.7;'>発話を順につなげて全文にし、<code>ai_extract</code>（項目の取り出し）と <code>ai_analyze_sentiment</code>（感情）にかけます。通話ごとの「困りごと・案内した解決策・後日対応の要否」が表になります。</div></div>
</div>

```python
def options(d):
    # Python の辞書を、AI 関数に渡す map(...) の列に変換する
    return F.create_map(*[F.lit(x) for kv in d.items() for x in kv])

call_schema = '''{
  "product":        {"type": "string",  "description": "話題になっている製品"},
  "issue":          {"type": "string",  "description": "お客様の困りごと（日本語で）"},
  "resolution":     {"type": "string",  "description": "オペレーターが案内した解決策（日本語で）"},
  "order_number":   {"type": "string",  "description": "注文番号。言っていなければ null"},
  "follow_up_needed": {"type": "boolean", "description": "後日の対応が必要か"}
}'''

full_text = transcribed.select(
    "file_name",
    F.array_join(
        F.transform(F.variant_get("transcription", "$.response.segments", "array<variant>"),
                    lambda s: F.concat(F.lit("[話者"), F.variant_get(s, "$.speaker_id", "string"), F.lit("] "),
                                       F.variant_get(s, "$.text", "string"))),
        "\n").alias("transcript"))

calls = full_text.select(
    "file_name", "transcript",
    F.call_function("ai_analyze_sentiment", F.col("transcript")).alias("sentiment"),
    F.variant_get(F.call_function("ai_extract", F.col("transcript"), F.lit(call_schema), options({"version": "2.1"})),
                  "$.response", "variant").alias("call_summary"))
display(calls)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div></div>
</div>

```python
result = segments

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'file_name': '録音ファイル名',
    'start_sec': '発話の開始（秒）',
    'end_sec': '発話の終了（秒）',
    'speaker': '話者ID（話し始めた順に 1, 2, ...）',
    'text': '書き起こしたテキスト',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.call_transcripts")
       .tableProperty("comment", '通話録音を ai_transcribe で文字に起こした結果（発話ごと）')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.call_transcripts"))
```

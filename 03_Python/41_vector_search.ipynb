```python
from databricks.ai_search.client import AISearchClient

client = AISearchClient()
source_table = f"{CATALOG}.{SCHEMA}.faq_articles"

# エンドポイント（検索を動かすサーバー）がなければ作る
if not client.endpoint_exists(VS_ENDPOINT):
    client.create_endpoint_and_wait(name=VS_ENDPOINT, endpoint_type="STANDARD")

# インデックスがなければ作る。あれば最新のテーブルの内容に同期する
if not client.index_exists(endpoint_name=VS_ENDPOINT, index_name=VS_INDEX):
    index = client.create_delta_sync_index_and_wait(
        endpoint_name=VS_ENDPOINT,
        index_name=VS_INDEX,
        source_table_name=source_table,
        pipeline_type="TRIGGERED",                      # 必要なときに同期する（CONTINUOUS なら常時同期）
        primary_key="article_id",
        embedding_source_column="body",                 # この列をベクトル化する
        embedding_model_endpoint_name=EMBEDDING_ENDPOINT,
    )
else:
    index = client.get_index(endpoint_name=VS_ENDPOINT, index_name=VS_INDEX)
    # テーブルを作り直した直後などは同期を受け付けないことがあるので、受け付けるまで待って再試行する
    import time
    for _ in range(40):
        try:
            index.sync()
            break
        except Exception as e:
            if "not ready to sync" not in str(e):
                raise
            time.sleep(15)

index.wait_until_ready()
print("✅ インデックスの準備ができました:", VS_INDEX)
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">文で検索する</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>query_text</code> に探したい文を渡すと、意味の近い順に返ります。「荷物」という語は FAQ にありませんが、「届かない」記事が見つかります。</div></div>
</div>

```python
import pandas as pd

query = "荷物がまだ来ないんですけど"
columns = ["article_id", "title", "category"]

def search(text, num_results=3, query_type=None):
    """AI Search で検索し、結果を Spark の DataFrame で返す"""
    r = index.similarity_search(query_text=text, columns=columns, num_results=num_results, query_type=query_type)
    names = [c["name"] for c in r["manifest"]["columns"]]
    return spark.createDataFrame(pd.DataFrame(r["result"].get("data_array", []), columns=names))

display(search(query))
```

```python
from pyspark.sql import functions as F

inquiries = spark.table(f"{CATALOG}.{SCHEMA}.inquiries").orderBy("inquiry_id").collect()
rows = []
for q in inquiries:
    top = search(q["body"], num_results=1).first()
    rows.append((q["inquiry_id"], q["body"], top["article_id"], top["title"], round(float(top["score"]), 3)))
matched = spark.createDataFrame(rows, "inquiry_id INT, body STRING, article_id STRING, faq_title STRING, score DOUBLE")
display(matched)
```

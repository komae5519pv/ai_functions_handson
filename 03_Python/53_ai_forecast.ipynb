```python
import time
import pandas as pd
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.sql import StatementParameterListItem, StatementState

w = WorkspaceClient()

def run_on_warehouse(sql, **params):
    """SQL ウェアハウス（00_config の WAREHOUSE_ID）で SQL を実行し、結果を pandas の DataFrame で返す"""
    r = w.statement_execution.execute_statement(
        warehouse_id=WAREHOUSE_ID, catalog=CATALOG, schema=SCHEMA, statement=sql, wait_timeout="50s",
        parameters=[StatementParameterListItem(name=k, value=str(v)) for k, v in params.items()] or None)
    while r.status.state in (StatementState.PENDING, StatementState.RUNNING):
        time.sleep(3)
        r = w.statement_execution.get_statement(r.statement_id)
    if r.status.state != StatementState.SUCCEEDED:
        raise RuntimeError(r.status.error.message)
    if not (r.manifest and r.manifest.schema and r.manifest.schema.columns):
        return None
    cols = [c.name for c in r.manifest.schema.columns]
    return pd.DataFrame((r.result.data_array if r.result else None) or [], columns=cols)
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>3 店舗の 2025 年 4 月〜2026 年 9 月（1 年半）の日次売上を作ります。右肩上がりのトレンドに、週末の波と年末商戦の山を重ねてあります。</div></div>

```python
from pyspark.sql import functions as F

start = F.lit("2025-04-01").cast("date")
dates = spark.range(548).select(F.date_add(start, F.col("id").cast("int")).alias("sales_date"))
stores = spark.createDataFrame([("新宿店", 1.00), ("梅田店", 0.80), ("博多店", 0.55)], "store STRING, scale DOUBLE")

base = (dates.crossJoin(stores)
        .withColumn("t", F.datediff("sales_date", start))
        .withColumn("weekend", F.when(F.dayofweek("sales_date").isin(1, 7), 1.35).otherwise(1.0))
        .withColumn("r", (F.abs(F.hash("store", "sales_date")) % 1000) / 1000.0))

# 季節の波: 年末（12/24 ごろ）に大きな山、夏（8/10 ごろ）に小さな山
doy = F.dayofyear("sales_date")
to_xmas = F.least(F.abs(doy - 358), 365 - F.abs(doy - 358))
base = base.withColumn("season", 1.0 + 0.32 * F.exp(-F.pow(to_xmas, 2) / (2 * 14 * 14))
                                     + 0.08 * F.exp(-F.pow(doy - 222, 2) / (2 * 18 * 18)))

def on(store, day):
    return (F.col("store") == store) & (F.col("sales_date") == F.lit(day).cast("date"))

event = (F.when(on("梅田店", "2026-08-15"), 0.35)    # システム障害でレジ停止
          .when(on("新宿店", "2026-07-20"), 1.90)    # 近隣イベントで急増
          .when(on("博多店", "2026-09-02"), 0.40)    # 台風で臨時休業（半日）
          .otherwise(1.0))

df = (base.select(
        "sales_date",
        "store",
        F.round(F.lit(2000000) * F.col("scale") * F.col("weekend") * F.col("season") * (1 + F.col("t") * 0.0015)
                * (0.93 + F.col("r") * 0.14) * event, -3).cast("bigint").alias("sales_amount"),
        F.round(480 * F.col("scale") * F.col("weekend") * (0.9 + F.col("r") * 0.2)).cast("int").alias("visitors"))
      .orderBy("store", "sales_date"))

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'sales_date'  : '営業日',
    'store'       : '店舗名',
    'sales_amount': '1日の売上金額（円）',
    'visitors'    : '1日の来店客数（人）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.daily_sales")
   .tableProperty("comment", '店舗別の日次売上（AI Functions デモ用の架空データ。2025年4月〜2026年9月。右肩上がりで年末に山がある。異常値を意図的に3日分入れてある）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.daily_sales"))
```

```python
from pyspark.sql import functions as F

horizon = "2026-12-31"      # 予測の期限（実績の最終日から 3 か月先まで）
holiday_region = "JP"

pdf = run_on_warehouse(f"""
SELECT *
FROM ai_forecast(
  TABLE(SELECT sales_date, store, sales_amount FROM daily_sales),
  horizon => '{horizon}', time_col => 'sales_date', value_col => 'sales_amount', group_col => 'store',
  holiday_region => '{holiday_region}', positive_only => true, version => '2')
""")
forecast = (spark.createDataFrame(pdf)
            .select(F.col("sales_date").cast("date"), "store",
                    *[F.col(c).cast("double") for c in ["sales_amount_forecast", "sales_amount_upper", "sales_amount_lower"]]))
display(forecast.orderBy("store", "sales_date"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">実績と予測をつなげて見る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>実績（2026 年 4〜9 月）と予測（10〜12 月）を1つの表にして、1本の線でつながるグラフにします。学習には 1 年半分の実績をすべて使い（前の年の年末の山を覚えさせるため）、グラフには直近の半年だけを出します。グラフにしやすいよう、線の種類（実績・予測・下限・上限）を <code>series</code> 列に、値を <code>amount</code> 列に入れた縦長の表にし、予測の線は実績の最終日（9/30）から始めています。</div></div>

```python
show_from = "2026-04-01"   # グラフに出す実績の開始日
last_actual = "2026-09-30" # 実績の最終日（予測の線の始点にする）

actual = (spark.table(f"{CATALOG}.{SCHEMA}.daily_sales")
          .where(F.col("sales_date") >= show_from)
          .select("sales_date", "store", F.col("sales_amount").cast("double").alias("amount")))

# 予測・下限・上限の線を、実績の最終日から始めてつなげる
bridge = (actual.where(F.col("sales_date") == last_actual)
          .crossJoin(spark.createDataFrame([("予測",), ("下限",), ("上限",)], "series STRING")))

# 予測（1 行を 予測・下限・上限 の 3 行に分ける）
predicted = forecast.select(
    "sales_date", "store",
    F.expr("stack(3, '予測', sales_amount_forecast, '下限', sales_amount_lower, '上限', sales_amount_upper) AS (series, amount)"))

df = (actual.withColumn("series", F.lit("実績"))
      .unionByName(bridge)
      .unionByName(predicted))
display(df.orderBy("store", "series", "sales_date"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">複数の値をまとめて予測する</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>value_col</code> に配列を渡すと、売上と来店客数を一度に予測できます。<code>prediction_interval_width</code> で予測区間の幅（ここでは 80%）も変えられます。</div></div>
</div>

```python
interval_width = 0.8   # 予測区間の幅
value_cols = ["sales_amount", "visitors"]

pdf = run_on_warehouse(f"""
SELECT * FROM ai_forecast(
  TABLE(SELECT sales_date, store, {", ".join(value_cols)} FROM daily_sales),
  horizon => '2026-10-15', time_col => 'sales_date', value_col => ARRAY({", ".join(f"'{c}'" for c in value_cols)}),
  group_col => 'store', prediction_interval_width => {interval_width}, version => '2')
""")
display(spark.createDataFrame(pdf).orderBy("store", "sales_date"))
```

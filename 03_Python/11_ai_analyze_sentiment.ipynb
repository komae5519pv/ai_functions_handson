```python
from pyspark.sql import functions as F

rows = [
    (1, 'ワイヤレスイヤホン AirTone S2', 5, '軽くて耳が痛くならない。バッテリーも一日もつので出張に最適です。'),
    (2, 'ロボット掃除機 CleanBot X', 1, '2週間で電源が入らなくなった。サポートに連絡しても返事が遅く最悪です。'),
    (3, 'ワイヤレスイヤホン AirTone S2', 3, '音質は良いが、ケースが安っぽく傷がつきやすい。値段相応かな。'),
    (4, '全自動コーヒーメーカー BrewMaster Pro', 3, '説明書がわかりにくく設定に1時間かかった。動き出してからは毎朝快適です。'),
    (5, 'ポータブル電源 PX-500', 4, '期待通りの性能。キャンプで2日使えた。配送も早かった。'),
    (6, 'ノートPCスタンド FlexLift', 2, '色が写真と全然違う。ぐらつきもあるので返品を考えています。'),
    (7, '全自動コーヒーメーカー BrewMaster Pro', 5, '毎日使っています。豆から挽けるので香りが段違い。もう手放せません！'),
    (8, 'ロボット掃除機 CleanBot X', 3, '普通です。特に良くも悪くもない。'),
]
df = spark.createDataFrame(rows, "review_id INT, product_name STRING, rating INT, review_text STRING")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'review_id'   : 'レビューID',
    'product_name': 'レビュー対象の商品名',
    'rating'      : '星の数（1〜5）',
    'review_text' : 'レビュー本文（お客様の原文）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.reviews")
   .tableProperty("comment", 'ECサイトの商品レビュー（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.reviews"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">まずは1文で試す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>文字列を直接渡すだけで動きます。</div></div>

```python
from pyspark.sql import functions as F

examples = ["配送が早くて梱包も丁寧でした。また利用します。",
            "届いた商品が壊れていて、問い合わせても返事がない。",
            "デザインは気に入ったが、思ったより音が大きい。"]
df = (spark.createDataFrame([(t,) for t in examples], "text STRING")
      .withColumn("sentiment", F.call_function("ai_analyze_sentiment", F.col("text"))))
display(df)
```

      <strong style="color: #00695c; font-size: 1.08em;">ここを見る</strong>
      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">良い点と悪い点が両方ある文は <code>mixed</code> になります。「良い / 悪い」の2択ではなく4種類で返るのがこの関数の特徴です。</div>
    </div>

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">テーブルの列にまとめてかける</div><div style='color:#555; margin-top:6px; line-height:1.7;'>列に対して呼ぶと、全行を一度に判定できます。星の数と並べてみます。</div></div>
</div>

```python
reviews = spark.table(f"{CATALOG}.{SCHEMA}.reviews")

# F.expr で SQL の関数をそのまま書く
df = reviews.withColumn("sentiment", F.expr("ai_analyze_sentiment(review_text)"))
display(df.orderBy("review_id"))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">集計して使う</div><div style='color:#555; margin-top:6px; line-height:1.7;'>判定結果は普通の文字列なので、そのまま集計できます。商品ごとにネガティブの件数を数えます。</div></div>
</div>

```python
judged = reviews.withColumn("sentiment", F.call_function("ai_analyze_sentiment", F.col("review_text")))

df = (judged.groupBy("product_name")
      .agg(F.count("*").alias("review_count"),
           F.count_if(F.col("sentiment") == "negative").alias("negative"),
           F.count_if(F.col("sentiment") == "mixed").alias("mixed"))
      .withColumn("negative_ratio", F.round(F.col("negative") / F.col("review_count"), 2))
      .orderBy(F.desc("negative"), F.desc("mixed")))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">6</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>AI の呼び出しは1行ごとに時間と料金がかかるので、結果は一度テーブルに保存して、集計やダッシュボードはそのテーブルから行います。</div></div>
</div>

```python
result = reviews.withColumn("sentiment", F.call_function("ai_analyze_sentiment", F.col("review_text")))

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'sentiment': 'AI が判定した感情（positive / negative / neutral / mixed）',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.reviews_sentiment")
       .tableProperty("comment", 'reviews に ai_analyze_sentiment の判定結果を付けたテーブル')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.reviews_sentiment"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルファイルを確認する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>ボリュームのファイルは <code>read_files</code> の <code>binaryFile</code> 形式で読むと、1ファイル1行で <code>content</code> 列（中身のバイナリ）が得られます。</div></div>

```python
from pyspark.sql import functions as F

def options(d):
    # Python の辞書を、AI 関数に渡す map(...) の列に変換する
    return F.create_map(*[F.lit(x) for kv in d.items() for x in kv])

invoices = spark.read.format("binaryFile").load(f"{VOL_ROOT}/invoices/")
display(invoices.select("path", "length", "modificationTime"))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">1つの PDF を読み取る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>戻り値は VARIANT です。<code>document.elements</code> に、本文・表・見出しなどの要素が順に並びます。</div></div>
</div>

```python
one = spark.read.format("binaryFile").load(f"{VOL_ROOT}/invoices/invoice_01.pdf")
df = one.select("path", F.call_function("ai_parse_document", F.col("content"), options({"version": "2.0"})).alias("parsed"))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">要素を1行ずつに展開する</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>F.variant_get(..., "array&lt;variant&gt;")</code> で配列として取り出し、<code>F.posexplode</code> で行に展開します。表は HTML で入っています。</div></div>
</div>

```python
docs = spark.read.format("binaryFile").load(f"{VOL_ROOT}/docs/")
parsed_docs = docs.select("path", F.call_function("ai_parse_document", F.col("content"), options({"version": "2.0"})).alias("doc"))

elements = parsed_docs.select(
    "path", F.posexplode(F.variant_get("doc", "$.document.elements", "array<variant>")).alias("element_no", "e"))
df = elements.select(
    "element_no",
    (F.variant_get("e", "$.bbox[0].page_id", "int") + 1).alias("page"),
    F.variant_get("e", "$.type", "string").alias("type"),
    F.variant_get("e", "$.content", "string").alias("content"),
    F.variant_get("e", "$.description", "string").alias("description"))
display(df.orderBy("element_no"))
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">読み取った結果から項目を取り出す（ai_extract と組み合わせる）</div><div style='color:#555; margin-top:6px; line-height:1.7;'>ai_parse_document の結果（VARIANT）は、そのまま <code>ai_extract</code> に渡せます。請求書 3 枚から、発行元・金額・明細を表にします。</div></div>

```python
invoice_schema = '''{
  "vendor":      {"type": "string",  "description": "請求書を発行した会社名"},
  "invoice_no":  {"type": "string",  "description": "請求番号"},
  "issue_date":  {"type": "string",  "description": "発行日（YYYY-MM-DD）"},
  "due_date":    {"type": "string",  "description": "支払期限（YYYY-MM-DD）"},
  "total_jpy":   {"type": "integer", "description": "税込の請求金額（円）"},
  "line_items": {
    "type": "array",
    "description": "明細の行",
    "items": {"type": "object", "properties": {
      "item":       {"type": "string"},
      "quantity":   {"type": "integer"},
      "unit_price": {"type": "integer"}
    }}
  }
}'''

parsed_invoices = invoices.select(
    "path", F.call_function("ai_parse_document", F.col("content"), options({"version": "2.0"})).alias("doc"))
df = parsed_invoices.select(
    F.regexp_extract("path", r"[^/]+$", 0).alias("file_name"),
    F.variant_get(
        F.call_function("ai_extract", F.col("doc"), F.lit(invoice_schema),
                        options({"version": "2.1", "instructions": "日本の請求書です。"})),
        "$.response", "variant").alias("invoice"))
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">6</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">画像（写真・スキャン）も同じように読める</div><div style='color:#555; margin-top:6px; line-height:1.7;'>画像ファイルも同じ関数で文字を読み取れます。レシートの写真を読み取ります。</div></div>
</div>

```python
images = spark.read.format("binaryFile").load(f"{VOL_ROOT}/images/")
df = (images
      .select(F.explode(F.variant_get(F.call_function("ai_parse_document", F.col("content"), options({"version": "2.0"})),
                                      "$.document.elements", "array<variant>")).alias("e"))
      .select(F.variant_get("e", "$.type", "string").alias("type"),
              F.variant_get("e", "$.content", "string").alias("content")))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">7</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">オプションで読み方を変える</div><div style='color:#555; margin-top:6px; line-height:1.7;'>読むページを絞る（<code>pageRange</code>）、図の説明を作らない（<code>descriptionElementTypes</code> を空にすると、図の多い文書で処理量と料金を抑えられます）などを指定できます。</div></div>
</div>

```python
page_range = "2"              # 読むページ（例: "1,3,5-10"）
description_types = ""        # 図の説明を作らない

parsed_p2 = docs.select(F.call_function(
    "ai_parse_document", F.col("content"),
    options({"version": "2.0", "pageRange": page_range, "descriptionElementTypes": description_types})).alias("doc"))
df = (parsed_p2.select(F.explode(F.variant_get("doc", "$.document.elements", "array<variant>")).alias("e"))
      .select(F.variant_get("e", "$.type", "string").alias("type"),
              F.variant_get("e", "$.content", "string").alias("content")))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">8</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">読み取り結果をテーブルに保存する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>読み取りは時間と料金がかかるので、結果は保存して使い回します。後続の <code>32_ai_prep_search</code> や <code>13_ai_extract</code> はこのテーブルから始められます。</div></div>
</div>

```python
all_files = (spark.read.format("binaryFile").option("recursiveFileLookup", "true").load(f"{VOL_ROOT}/")
             .where(F.col("path").rlike(r"\.(pdf|png|jpg|jpeg)$")))
result = all_files.select(
    "path", F.call_function("ai_parse_document", F.col("content"), options({"version": "2.0"})).alias("parsed"))

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'path': '元のファイルのパス',
    'parsed': 'ai_parse_document の戻り値（VARIANT）。document.elements に要素の配列',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.documents_parsed")
       .tableProperty("comment", 'ボリュームの PDF・画像を ai_parse_document で読み取った結果')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.documents_parsed"))
```

```python
from pyspark.sql import functions as F

def options(d):
    # Python の辞書を、AI 関数に渡す map(...) の列に変換する
    return F.create_map(*[F.lit(x) for kv in d.items() for x in kv])

invoices = spark.read.format("binaryFile").load(f"{VOL_ROOT}/invoices/")
display(invoices.select("path", "length", "modificationTime"))
```

```python
docs = spark.read.format("binaryFile").load(f"{VOL_ROOT}/docs/")
parsed_docs = docs.select("path", F.call_function("ai_parse_document", F.col("content"), options({"version": "2.0"})).alias("doc"))

elements = parsed_docs.select(
    "path", F.posexplode(F.variant_get("doc", "$.document.elements", "array<variant>")).alias("element_no", "e"))
df = elements.select(
    "element_no",
    (F.variant_get("e", "$.bbox[0].page_id", "int") + 1).alias("page"),
    F.variant_get("e", "$.type", "string").alias("type"),
    F.variant_get("e", "$.content", "string").alias("content"),
    F.variant_get("e", "$.description", "string").alias("description"))
display(df.orderBy("element_no"))
```

```python
invoice_schema = '''{
  "vendor":      {"type": "string",  "description": "請求書の発行元の会社名（宛先の「御中」の会社ではない）"},
  "invoice_no":  {"type": "string",  "description": "請求番号"},
  "issue_date":  {"type": "string",  "description": "発行日（YYYY-MM-DD）"},
  "due_date":    {"type": "string",  "description": "支払期限（YYYY-MM-DD）"},
  "total_jpy":   {"type": "integer", "description": "税込の請求金額（円）"},
  "line_items": {
    "type": "array",
    "description": "明細の行",
    "items": {"type": "object", "properties": {
      "item":       {"type": "string"},
      "quantity":   {"type": "integer"},
      "unit_price": {"type": "integer"}
    }}
  }
}'''

parsed_invoices = invoices.select(
    "path", F.call_function("ai_parse_document", F.col("content"), options({"version": "2.0"})).alias("doc"))
df = parsed_invoices.select(
    F.regexp_extract("path", r"[^/]+$", 0).alias("file_name"),
    F.variant_get(
        F.call_function("ai_extract", F.col("doc"), F.lit(invoice_schema),
                        options({"version": "2.1", "mode": "precision",
                                 "instructions": "日本の請求書から、請求した側（発行元）の情報を取り出します。請求書の上部で「御中」が付いている会社は宛先（請求された側）なので、vendor にしないでください。発行元は、登録番号・住所・電話番号が書かれている会社です。"})),
        "$.response", "variant").alias("invoice"))
display(df)
```

```python
from pyspark.sql import functions as F

rows = [
    ('株式会社ミナト物流', 'V001'),
    ('グリーンオフィス株式会社', 'V002'),
    ('クラウドワークス・システムズ合同会社', 'V003'),
]
df = spark.createDataFrame(rows, "vendor_name STRING, vendor_code STRING")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'vendor_name': '取引先の会社名',
    'vendor_code': '取引先コード',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.vendor_master")
   .tableProperty("comment", '取引先マスタ（AI Functions デモ用の架空データ。請求書の発行元の確かめに使う）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.vendor_master"))
```

```python
from pyspark.sql import functions as F

check_schema = '''{"vendor": {"type": "string", "description": "請求書の発行元の会社名（宛先の「御中」の会社ではない）"}, "total_jpy": {"type": "integer", "description": "税込の請求金額（円）"}, "line_items": {"type": "array", "items": {"type": "object", "properties": {"item": {"type": "string"}, "quantity": {"type": "integer"}, "unit_price": {"type": "integer"}}}}}'''
check_instructions = "日本の請求書から、請求した側（発行元）の情報を取り出します。請求書の上部で「御中」が付いている会社は宛先（請求された側）なので、vendor にしないでください。"

# ① まず、組み込みの ai_extract で全件を読む
checked = (parsed_invoices
    .withColumn("file_name", F.regexp_extract("path", r"[^/]+$", 0))
    .withColumn("r", F.variant_get(
        F.call_function("ai_extract", F.col("doc"), F.lit(check_schema),
                        options({"version": "2.1", "mode": "precision", "instructions": check_instructions})),
        "$.response", "variant"))
    .select("file_name", "doc",
            F.variant_get("r", "$.vendor.value", "string").alias("vendor"),
            F.variant_get("r", "$.total_jpy.value", "bigint").alias("total_jpy"),
            F.expr("aggregate(from_json(to_json(r:line_items), 'ARRAY<STRUCT<quantity: STRUCT<value: BIGINT>, unit_price: STRUCT<value: BIGINT>>>'), "
                   "0L, (acc, x) -> acc + x.quantity.value * x.unit_price.value)").alias("subtotal")))

# ② SQL のルールで確かめる（LLM は呼ばない）
master = spark.table(f"{CATALOG}.{SCHEMA}.vendor_master")
judged = (checked.join(master, checked.vendor == master.vendor_name, "left")
    .withColumn("rule_required", F.col("vendor").isNotNull() & F.col("total_jpy").isNotNull())   # 必須の項目が空でない
    .withColumn("rule_total", F.round(F.col("subtotal") * 1.1) == F.col("total_jpy"))             # 明細の合計 × 1.1 ＝ 請求金額
    .withColumn("rule_vendor_master", F.col("vendor_name").isNotNull())                            # 発行元が取引先マスタにある
    .withColumn("passed", F.col("rule_required") & F.col("rule_total") & F.col("rule_vendor_master")))

# ③ ルールに引っかかった行だけを、ai_query で日本語に強いモデルに読み直させる
reread = judged.where(~F.col("passed")).select(
    "file_name",
    F.expr(f"""ai_query('{LLM_ENDPOINT}',
        concat('次は請求書を読み取った結果です。請求した側（発行元）の会社名と、税込の請求金額を取り出してください。上部で「御中」が付いている会社は宛先なので、発行元にしないでください。\\n\\n', to_json(doc:document.elements)),
        responseFormat => 'STRUCT<result: STRUCT<vendor: STRING, total_jpy: BIGINT>>')""").alias("reread_result"))

display(judged.join(reread, "file_name", "left")
        .select("file_name", "vendor", "total_jpy", "subtotal", "rule_required", "rule_total", "rule_vendor_master", "passed", "reread_result")
        .orderBy("file_name"))
```

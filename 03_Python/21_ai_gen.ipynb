```python
from pyspark.sql import functions as F

rows = [
    ('P001', 'ポータブル電源 PX-500', '家電', '容量512Wh / AC出力500W / 重量6.2kg / ソーラー充電対応', 69800),
    ('P002', 'ワイヤレスイヤホン AirTone S2', 'オーディオ', 'ノイズキャンセリング / 最大30時間再生 / 防水IPX4', 14800),
    ('P003', '電動アシスト自転車 RideOn 26', 'モビリティ', '26インチ / 航続距離80km / 7段変速', 128000),
    ('P004', '全自動コーヒーメーカー BrewMaster Pro', 'キッチン家電', 'ミル付き / 最大10杯 / タイマー予約', 32800),
    ('P005', 'ノートPCスタンド FlexLift', 'オフィス用品', 'アルミ製 / 高さ6段階 / 折りたたみ式', 4980),
    ('P006', 'ロボット掃除機 CleanBot X', '生活家電', 'マッピング機能 / 自動ゴミ収集 / 水拭き対応', 59800),
]
df = spark.createDataFrame(rows, "product_id STRING, product_name STRING, category STRING, spec STRING, price_jpy INT")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'product_id'  : '商品ID',
    'product_name': '自社の正式な商品名',
    'category'    : '商品カテゴリ',
    'spec'        : '主な仕様（社内の商品マスタに登録された短いメモ）',
    'price_jpy'   : '税込販売価格（円）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.products")
   .tableProperty("comment", '自社の商品マスタ（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.products"))
```

```python
from pyspark.sql import functions as F

prompt = "秋のキャンプ用品セールのメール件名を、30文字以内で3案、箇条書きで出してください。"
df = spark.range(1).select(F.call_function("ai_gen", F.lit(prompt)).alias("subject_ideas"))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">列の値を指示文に埋め込んで、行ごとに作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>concat</code> で商品名や仕様を指示文に入れると、商品ごとに紹介文ができます。</div></div>
</div>

```python
instruction = ("あなたは EC サイトのコピーライターです。次の商品の紹介文を 80 文字以内の日本語で1つだけ書いてください。"
               "誇張表現は使わず、仕様に書かれていないことは書かないでください。")

products = spark.table(f"{CATALOG}.{SCHEMA}.products")
prompt = F.concat(F.lit(instruction),
                  F.lit("\n商品名: "), F.col("product_name"),
                  F.lit("\nカテゴリ: "), F.col("category"),
                  F.lit("\n仕様: "), F.col("spec"))
df = products.select("product_id", "product_name", "spec", F.call_function("ai_gen", prompt).alias("catch_copy"))
display(df)
```

<div style="border-left: 4px solid #009688; background: #e0f2f1; padding: 14px 20px; border-radius: 4px; margin: 14px 0;">
  <div style="display: flex; align-items: flex-start; gap: 12px;">
    <span style="font-size: 22px;">💡</span>
    <div>
      <strong style="color: #00695c; font-size: 1.08em;">ai_gen と ai_query の使い分け</strong>
      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">ai_gen はモデルを選べない代わりに、指示文を渡すだけで使えます。<b>モデルを選びたい・出力を JSON の形に固定したい・画像を読ませたい</b>ときは <code>22_ai_query</code> を使います。</div>
    </div>
  </div>
</div>

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>生成した文章は保存して、人が確認・修正してから公開します。</div></div>
</div>

```python
result = products.withColumn("catch_copy", F.call_function("ai_gen", prompt))

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'catch_copy': 'AI が作った紹介文の下書き（公開前に人が確認する）',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.products_copy")
       .tableProperty("comment", 'products に ai_gen で作った紹介文の下書きを付けたテーブル')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.products_copy"))
```

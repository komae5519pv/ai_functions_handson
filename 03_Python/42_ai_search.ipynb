```python
from pyspark.sql import functions as F
from datetime import datetime

rows = [
    (1, datetime(2026, 10, 14, 9, 12), 'メール', '注文番号 A-10231 の商品がまだ届きません。10月10日に注文し、到着予定は14日でした。明日のイベントで使うので至急確認してください。山田花子 090-1111-2222'),
    (2, datetime(2026, 10, 14, 10, 40), 'Webフォーム', '請求書の金額が見積と違います。見積では税込132,000円でしたが、届いた請求書は145,200円になっています。確認をお願いします。経理部 佐藤 sato@example.co.jp'),
    (3, datetime(2026, 10, 14, 11, 5), 'チャット', 'ポータブル電源PX-500を使っていたら焦げ臭いにおいがして、本体が異常に熱くなりました。怖くて使えません。どうすればいいですか。'),
    (4, datetime(2026, 10, 14, 13, 22), 'Webフォーム', '会員登録したメールアドレスを変更したいです。新しいアドレスは k.tanaka@example.com です。手順を教えてください。'),
    (5, datetime(2026, 10, 14, 14, 1), 'メール', '先日購入したコーヒーメーカー、とても使いやすく満足しています。店頭スタッフの方の説明も丁寧でした。ありがとうございました。'),
    (6, datetime(2026, 10, 14, 15, 30), 'メール', '法人向けにノートPCを50台まとめて購入したいのですが、ボリュームディスカウントはありますか。来月中に導入したいです。株式会社サンプル 鈴木一郎 03-5555-1234'),
    (7, datetime(2026, 10, 14, 16, 45), 'Webフォーム', '返品したいです。サイズが合いませんでした。未開封です。送り先は 大阪府大阪市北区梅田1-2-3 の高橋です。'),
    (8, datetime(2026, 10, 14, 18, 10), 'チャット', 'アプリにログインできません。パスワードを再設定してもエラーになります。昨日から何度も試していて困っています。'),
]
df = spark.createDataFrame(rows, "inquiry_id INT, received_at TIMESTAMP, channel STRING, body STRING")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'inquiry_id' : '問い合わせID',
    'received_at': '受付日時',
    'channel'    : '受付チャネル（メール / Webフォーム / チャット）',
    'body'       : '問い合わせ本文（お客様が書いた原文。個人情報を含む）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.inquiries")
   .tableProperty("comment", 'カスタマーサポートに届いた問い合わせ（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries"))
```

```python
import json
from pyspark.sql import functions as F

question = "返品するとき、送料は誰が払うの？"
# 情報源の設定（Python の辞書で書いて JSON にする）
knowledge_sources = json.dumps([{
    "type": "vector_search",
    "config": {"index_name": VS_INDEX, "text_col": "body", "doc_uri_col": "url"},
}])
sources = F.parse_json(F.lit(knowledge_sources))

df = spark.range(1).select(F.call_function("ai_search", F.lit(question), sources).alias("result"))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">答えと出典を列に取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>F.variant_get(結果, "$.answer", "string")</code> で答え、<code>$.document[0].doc_uri</code> で一番関連の高い文書の URL を取り出します。<code>generate_citations</code> を付けると、答えの根拠になった文書の番号（<code>citations</code>）も返ります。</div></div>
</div>

```python
def options(d):
    # Python の辞書を、AI 関数に渡す map(...) の列に変換する
    return F.create_map(*[F.lit(x) for kv in d.items() for x in kv])

q = "ポータブル電源が熱くなって変なにおいがする。どうしたらいい？"
r = spark.range(1).select(
    F.call_function("ai_search", F.lit(q), sources, F.lit(None), options({"generate_citations": "true"})).alias("result"))
df = r.select(
    F.variant_get("result", "$.answer", "string").alias("answer"),
    F.variant_get("result", "$.document[0].doc_uri", "string").alias("top_source"),
    F.variant_get("result", "$.citations", "variant").alias("citations"),
    F.variant_get("result", "$.document", "variant").alias("documents"))
display(df)
```

```python
instructions = "カスタマーサポートの FAQ から、お客様の困りごとに直接答える記事を探してください。安全に関わる内容を最優先してください。"

inquiries = spark.table(f"{CATALOG}.{SCHEMA}.inquiries")
answered = inquiries.withColumn("r", F.call_function("ai_search", F.col("body"), sources, F.lit(instructions)))
df = answered.select(
    "inquiry_id", "body",
    F.variant_get("r", "$.answer", "string").alias("suggested_reply"),
    F.variant_get("r", "$.document[0].doc_uri", "string").alias("source_url"))
display(df.orderBy("inquiry_id"))
```

```python
result = df

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'suggested_reply': 'FAQ をもとに AI が作った回答の下書き',
    'source_url': '回答の根拠として最も関連の高かった FAQ 記事の URL',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.inquiries_suggested_reply")
       .tableProperty("comment", 'inquiries の各問い合わせに、ai_search で FAQ を根拠にした回答案を付けたテーブル')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.inquiries_suggested_reply"))
```

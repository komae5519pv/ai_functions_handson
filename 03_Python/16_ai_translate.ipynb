<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">2</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>日本語の商品説明 3 件と、英語・スペイン語のレビュー 2 件を作ります。</div></div>
</div>

```python
from pyspark.sql import functions as F

rows = [
    ('P001', 'ja', '容量512Whの大容量ポータブル電源。キャンプや停電時の備えに。ソーラーパネルからも充電できます。'),
    ('P002', 'ja', 'ノイズキャンセリング搭載のワイヤレスイヤホン。最大30時間再生で、雨の日でも安心の防水仕様です。'),
    ('P004', 'ja', '豆から挽いて淹れる全自動コーヒーメーカー。タイマー予約で、朝起きたら淹れたてのコーヒーが飲めます。'),
    ('R101', 'en', 'Great sound quality, but the charging case feels a bit cheap.'),
    ('R102', 'es', 'La cafetera es fácil de usar y el café sale delicioso.'),
]
df = spark.createDataFrame(rows, "item_id STRING, source_lang STRING, text STRING")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'item_id'    : '商品ID（P）またはレビューID（R）',
    'source_lang': '原文の言語コード',
    'text'       : '原文',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.product_descriptions")
   .tableProperty("comment", '商品説明文と、海外のお客様からのレビュー（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.product_descriptions"))
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">3</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">日本語 → 英語に翻訳する</div></div>
</div>

```python
from pyspark.sql import functions as F

descriptions = spark.table(f"{CATALOG}.{SCHEMA}.product_descriptions")
df = (descriptions.where(F.col("source_lang") == "ja")
      .select("item_id", "text", F.call_function("ai_translate", F.col("text"), F.lit("en")).alias("text_en")))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">4</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">複数の言語に一度に訳す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>列を並べれば、1回の処理で複数の言語に訳せます。</div></div>
</div>

```python
target_langs = {"english": "en", "spanish": "es", "thai": "th"}   # 列名: 言語コード

df = (descriptions.where(F.col("source_lang") == "ja")
      .select("item_id", "text",
              *[F.call_function("ai_translate", F.col("text"), F.lit(code)).alias(name) for name, code in target_langs.items()]))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">海外のレビューを英語にそろえる</div><div style='color:#555; margin-top:6px; line-height:1.7;'>原文の言語がばらばらでも、翻訳先を1つに決めれば、まとめて読んだり分析したりできます。</div></div>
</div>

```python
df = (descriptions.where(F.col("source_lang") != "ja")
      .withColumn("text_en", F.expr("ai_translate(text, 'en')")))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">6</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">日本語に訳したいときは ai_query を使う</div><div style='color:#555; margin-top:6px; line-height:1.7;'>日本語は ai_translate の公式の翻訳先に入っていないため、モデルを選べる <code>ai_query</code> で翻訳します（モデルは <code>00_config</code> の <code>LLM_ENDPOINT</code>）。</div></div>
</div>

```python
prompt = F.concat(F.lit("次の文章を自然な日本語に翻訳してください。訳文だけを出力してください。\n\n"), F.col("text"))
df = (descriptions.where(F.col("source_lang") != "ja")
      .select("item_id", "text", F.call_function("ai_query", F.lit(LLM_ENDPOINT), prompt).alias("text_ja")))
display(df)
```

<div style="display:flex; align-items:flex-start; gap:14px; margin: 26px 0 8px 0; padding-bottom:10px; border-bottom: 2px solid #e3eaf3;">
  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">7</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div></div>
</div>

```python
result = (descriptions.where(F.col("source_lang") == "ja")
          .select("item_id", F.col("text").alias("text_ja"),
                  F.call_function("ai_translate", F.col("text"), F.lit("en")).alias("text_en")))

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'text_ja': '日本語の原文',
    'text_en': 'AI が英訳した文章',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.product_descriptions_en")
       .tableProperty("comment", 'product_descriptions の日本語の商品説明を ai_translate で英訳したテーブル')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.product_descriptions_en"))
```

```python
from pyspark.sql import functions as F

rows = [
    ('P001', 'ポータブル電源 PX-500', '家電', '容量512Wh / AC出力500W / 重量6.2kg / ソーラー充電対応', 69800),
    ('P002', 'ワイヤレスイヤホン AirTone S2', 'オーディオ', 'ノイズキャンセリング / 最大30時間再生 / 防水IPX4', 14800),
    ('P003', '電動アシスト自転車 RideOn 26', 'モビリティ', '26インチ / 航続距離80km / 7段変速', 128000),
    ('P004', '全自動コーヒーメーカー BrewMaster Pro', 'キッチン家電', 'ミル付き / 最大10杯 / タイマー予約', 32800),
    ('P005', 'ノートPCスタンド FlexLift', 'オフィス用品', 'アルミ製 / 高さ6段階 / 折りたたみ式', 4980),
    ('P006', 'ロボット掃除機 CleanBot X', '生活家電', 'マッピング機能 / 自動ゴミ収集 / 水拭き対応', 59800),
]
df = spark.createDataFrame(rows, "product_id STRING, product_name STRING, category STRING, spec STRING, price_jpy INT")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'product_id'  : '商品ID',
    'product_name': '自社の正式な商品名',
    'category'    : '商品カテゴリ',
    'spec'        : '主な仕様（社内の商品マスタに登録された短いメモ）',
    'price_jpy'   : '税込販売価格（円）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.products")
   .tableProperty("comment", '自社の商品マスタ（AI Functions デモ用の架空データ）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.products"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">作りたい列を定義して足す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>スキーマに作りたい列を書きます。<code>instructions</code> で「日本語で」と指定すると日本語で返ります（指定しないと英語で返ることがあります）。</div></div>

```python
from pyspark.sql import functions as F

def options(d):
    # Python の辞書を、AI 関数に渡す map(...) の列に変換する
    return F.create_map(*[F.lit(x) for kv in d.items() for x in kv])

schema = '''{
  "target_customer": {"type": "string", "description": "想定する主な購入者"},
  "use_scenes":      {"type": "array", "items": {"type": "string"}, "description": "主な利用シーン（3つまで）"},
  "gift_suitable":   {"type": "boolean", "description": "贈り物に向いているか"},
  "price_range":     {"type": "enum", "labels": ["低価格帯", "中価格帯", "高価格帯"], "description": "このカテゴリの中での価格帯"}
}'''

products = spark.table(f"{CATALOG}.{SCHEMA}.products")
content = F.concat_ws(" / ", "product_name", "category", "spec")
df = products.select(
    "product_id", "product_name", "spec",
    F.call_function("ai_enrich", content, F.lit(schema), F.lit(None),
                    options({"instructions": "家電・日用品の EC サイトの商品です。日本語で答えてください。"})).alias("enriched"))
display(df)
```

      <strong style="color: #00695c; font-size: 1.08em;">ここを見る</strong>
      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">各列に <code>value</code>（値）と <code>rationale</code>（なぜその値にしたか）が入っています。理由が付くので、人が確認するときの手がかりになります。</div>
    </div>
  </div>

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">VARIANT から列に取り出す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>理由が不要なら <code>'enableRationale','false'</code> にすると <code>value</code> だけになります。配列の列は、要素ごとに <code>value</code> を取り出します。</div></div>
</div>

```python
schema3 = '''{
  "target_customer": {"type": "string", "description": "想定する主な購入者"},
  "use_scenes":      {"type": "array", "items": {"type": "string"}, "description": "主な利用シーン（3つまで）"},
  "gift_suitable":   {"type": "boolean", "description": "贈り物に向いているか"}
}'''

enriched = products.withColumn(
    "enriched",
    F.call_function("ai_enrich", content, F.lit(schema3), F.lit(None),
                    options({"instructions": "日本語で答えてください。", "enableRationale": "false"})))
df = enriched.select(
    "product_id", "product_name",
    F.variant_get("enriched", "$.response.target_customer.value", "string").alias("target_customer"),
    F.transform(F.variant_get("enriched", "$.response.use_scenes", "array<variant>"),
                lambda x: F.variant_get(x, "$.value", "string")).alias("use_scenes"),
    F.variant_get("enriched", "$.response.gift_suitable.value", "boolean").alias("gift_suitable"))
display(df)
```

  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">5</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">Web 検索を根拠にする</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>knowledge_sources</code> に <code>web_search</code> を指定すると、Web を検索した結果をもとに値を埋め、出典の URL を <code>metadata.sources</code> に返します。<code>allowed_domains</code> で参照するサイトを絞ることもできます。</div></div>
</div>

```python
company_schema = '''{
  "industry":          {"type": "string", "description": "主な業種"},
  "headquarters_city": {"type": "string", "description": "本社がある都市"},
  "founded_year":      {"type": "integer", "description": "設立した年"}
}'''
knowledge_sources = '[{"type": "web_search", "config": {}}]'

companies = spark.createDataFrame([("Databricks",), ("Apache Software Foundation",)], "company STRING")
df = companies.select(
    "company",
    F.call_function("ai_enrich", F.col("company"), F.lit(company_schema), F.parse_json(F.lit(knowledge_sources)),
                    options({"instructions": "公開情報で確認できない値は null にしてください。日本語で答えてください。",
                             "enableRationale": "false"})).alias("enriched"))
display(df)
```

    <span style="font-size: 22px;">💡</span>
    <div>
      <strong style="color: #00695c; font-size: 1.08em;">社内の文書を根拠にする</strong>
      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;"><code>{"type": "vector_search", "config": {"index_name": "...", "text_col": "...", "doc_uri_col": "..."}}</code> を指定すると、AI Search のインデックス（社内の FAQ やマニュアルなど）を根拠にできます。インデックスの作り方は <code>41_vector_search</code> を参照してください。</div>

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>Web 検索や AI の推論は毎回結果が変わることがあるので、一度保存したものを使います。</div></div>

```python
result = products.withColumn(
    "enriched",
    F.call_function("ai_enrich", content, F.lit(schema3), F.lit(None), options({"instructions": "日本語で答えてください。"})))

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'enriched': 'ai_enrich の戻り値（VARIANT）。response に列ごとの value と rationale',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.products_enriched")
       .tableProperty("comment", 'products に ai_enrich で想定顧客・利用シーン・贈答向きかを足したテーブル')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.products_enriched"))
```

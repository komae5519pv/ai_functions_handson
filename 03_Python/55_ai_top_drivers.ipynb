  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>2026 年 4〜9 月の注文明細（約 5,600 件）を作ります。<b>7〜9 月は「関西 × アプリ」の注文だけが大きく減る</b>ように作ってあります。これを関数が見つけられるかを確かめます。</div></div>

```python
from pyspark.sql import functions as F

def pick(values, salt):
    # hash で決まった位置の値を選ぶ（SQL 版と同じ値になる）
    idx = (F.abs(F.hash(F.col("id"), F.lit(salt))) % len(values)).cast("int") + 1
    return F.element_at(F.array(*[F.lit(v) for v in values]), idx)

g = spark.range(6000).select(
    F.col("id").alias("order_id"),
    F.date_add(F.lit("2026-04-01").cast("date"), (F.abs(F.hash(F.col("id"), F.lit("d"))) % 183).cast("int")).alias("order_date"),
    pick(["関東", "関西", "九州", "東北"], "r").alias("region"),
    pick(["店舗", "Web", "アプリ"], "c").alias("channel"),
    pick(["家電", "オーディオ", "キッチン家電", "オフィス用品"], "k").alias("category"),
    pick(["新規", "リピーター"], "u").alias("customer_type"),
    (3000 + F.abs(F.hash(F.col("id"), F.lit("a"))) % 27000).cast("bigint").alias("amount"),
    (F.abs(F.hash(F.col("id"), F.lit("x"))) % 10).alias("drop_key"))

# 7月以降の「関西 × アプリ」は 70% の注文を間引く
dropped = ((F.col("order_date") >= F.lit("2026-07-01").cast("date")) & (F.col("region") == "関西")
           & (F.col("channel") == "アプリ") & (F.col("drop_key") < 7))
df = g.where(~dropped).drop("drop_key")

# 列の説明（コメント）は、列のメタデータに入れて保存する
column_comments = {
    'order_id'     : '注文ID',
    'order_date'   : '注文日',
    'region'       : 'お届け先の地域',
    'channel'      : '注文チャネル（店舗 / Web / アプリ）',
    'category'     : '商品カテゴリ',
    'customer_type': '会員区分（新規 / リピーター）',
    'amount'       : '注文金額（円）',
}
df = df.select([F.col(c).alias(c, metadata={"comment": column_comments[c]}) for c in column_comments])

(df.writeTo(f"{CATALOG}.{SCHEMA}.orders")
   .tableProperty("comment", '注文明細（AI Functions デモ用の架空データ。7〜9月は関西×アプリの売上が落ちるように作ってある）')
   .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.orders"))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">4〜6月 と 7〜9月 の売上の差の要因を探す</div><div style='color:#555; margin-top:6px; line-height:1.7;'>比較元（4〜6月）を FALSE、比較先（7〜9月）を TRUE にした列 <code>is_test</code> を作って渡します。</div></div>
</div>

```python
from pyspark.sql import functions as F

test_start = "2026-07-01"   # この日以降を比較先にする
dimensions = ["region", "channel", "category", "customer_type"]

orders = spark.table(f"{CATALOG}.{SCHEMA}.orders")
input_df = orders.select(*dimensions, "amount", (F.col("order_date") >= F.lit(test_start).cast("date")).alias("is_test"))

def top_drivers(src_df, metric="amount", aggregation="sum"):
    # テーブル値関数なので spark.sql に DataFrame を渡して呼ぶ
    dims = ", ".join(f"'{d}'" for d in dimensions)
    return spark.sql(
        f"SELECT * FROM ai_top_drivers(input => TABLE({{src}}), metric => '{metric}', is_test => 'is_test', "
        f"dimensions => ARRAY({dims}), aggregation => '{aggregation}')",
        src=src_df)

drivers = top_drivers(input_df)
display(drivers.withColumn("relative_change", F.round("relative_change", 3))
               .withColumn("support", F.round("support", 3))
               .orderBy(F.desc(F.abs("change"))).limit(15))
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">件数（注文数）の差で見る</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>aggregation => 'count'</code> にすると件数の差を調べられます。数える列には、定数の列（<code>1 AS order_count</code>）を使うのが定番です。</div></div>

```python
count_input = input_df.drop("amount").withColumn("order_count", F.lit(1))
count_drivers = top_drivers(count_input, metric="order_count", aggregation="count")
display(count_drivers.orderBy(F.desc(F.abs("change"))).limit(10))
```

```python
result = drivers

# 足した列の説明（コメント）を、列のメタデータに入れて保存する
added_comments = {
    'contributor': '切り口（列=値 の配列。2つなら組み合わせ）',
    'change': '比較先 − 比較元 の売上の差（円）',
}
result = result.select([F.col(c).alias(c, metadata={"comment": added_comments[c]}) if c in added_comments else F.col(c)
                        for c in result.columns])
(result.writeTo(f"{CATALOG}.{SCHEMA}.sales_change_drivers")
       .tableProperty("comment", 'orders の 4〜6月→7〜9月 の売上の差を ai_top_drivers で要因分解した結果')
       .createOrReplace())

display(spark.table(f"{CATALOG}.{SCHEMA}.sales_change_drivers"))
```

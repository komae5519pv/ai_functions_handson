```sql
CREATE OR REPLACE TABLE inquiries_faq_match
COMMENT 'inquiries の各問い合わせに、vector_search で最も近い FAQ 記事を付けたテーブル'
AS
SELECT i.inquiry_id, i.body, v.article_id, v.title AS faq_title, v.search_score
FROM inquiries AS i,
LATERAL vector_search(index => vs_index, query_text => i.body, num_results => 1) AS v;

ALTER TABLE inquiries_faq_match ALTER COLUMN article_id   COMMENT '最も近い FAQ 記事のID';
ALTER TABLE inquiries_faq_match ALTER COLUMN search_score COMMENT '検索スコア（高いほど近い）';

SELECT * FROM inquiries_faq_match ORDER BY inquiry_id;
```

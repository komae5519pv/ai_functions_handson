```sql
CREATE OR REPLACE TABLE products
COMMENT '自社の商品マスタ（AI Functions デモ用の架空データ）'
AS
SELECT * FROM VALUES
  ('P001', 'ポータブル電源 PX-500',                 '家電',         '容量512Wh / AC出力500W / 重量6.2kg / ソーラー充電対応', 69800),
  ('P002', 'ワイヤレスイヤホン AirTone S2',         'オーディオ',   'ノイズキャンセリング / 最大30時間再生 / 防水IPX4',      14800),
  ('P003', '電動アシスト自転車 RideOn 26',          'モビリティ',   '26インチ / 航続距離80km / 7段変速',                     128000),
  ('P004', '全自動コーヒーメーカー BrewMaster Pro', 'キッチン家電', 'ミル付き / 最大10杯 / タイマー予約',                    32800),
  ('P005', 'ノートPCスタンド FlexLift',             'オフィス用品', 'アルミ製 / 高さ6段階 / 折りたたみ式',                   4980),
  ('P006', 'ロボット掃除機 CleanBot X',             '生活家電',     'マッピング機能 / 自動ゴミ収集 / 水拭き対応',            59800)
AS t(product_id, product_name, category, spec, price_jpy);

ALTER TABLE products ALTER COLUMN product_id COMMENT '商品ID';
ALTER TABLE products ALTER COLUMN product_name COMMENT '自社の正式な商品名';
ALTER TABLE products ALTER COLUMN category COMMENT '商品カテゴリ';
ALTER TABLE products ALTER COLUMN spec COMMENT '主な仕様（社内の商品マスタに登録された短いメモ）';
ALTER TABLE products ALTER COLUMN price_jpy COMMENT '税込販売価格（円）';

SELECT * FROM products;
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">列の値を指示文に埋め込んで、行ごとに作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>concat</code> で商品名や仕様を指示文に入れると、商品ごとに紹介文ができます。</div></div>

```sql
SELECT
  product_id,
  product_name,
  spec,
  ai_gen(concat(
    'あなたは EC サイトのコピーライターです。次の商品の紹介文を 80 文字以内の日本語で1つだけ書いてください。',
    '誇張表現は使わず、仕様に書かれていないことは書かないでください。',
    '\n商品名: ', product_name,
    '\nカテゴリ: ', category,
    '\n仕様: ', spec
  )) AS catch_copy
FROM products
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>生成した文章は保存して、人が確認・修正してから公開します。</div></div>

```sql
CREATE OR REPLACE TABLE products_copy
COMMENT 'products に ai_gen で作った紹介文の下書きを付けたテーブル'
AS
SELECT
  *,
  ai_gen(concat(
    '次の商品の紹介文を 80 文字以内の日本語で1つだけ書いてください。仕様に書かれていないことは書かないでください。',
    '\n商品名: ', product_name, '\n仕様: ', spec
  )) AS catch_copy
FROM products;

ALTER TABLE products_copy ALTER COLUMN catch_copy COMMENT 'AI が作った紹介文の下書き（公開前に人が確認する）';

SELECT * FROM products_copy;
```

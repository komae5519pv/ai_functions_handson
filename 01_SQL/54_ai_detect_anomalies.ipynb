  <div style="min-width:36px; height:36px; line-height:36px; border-radius:8px; background:#1565c0; color:#fff; font-weight:700; text-align:center; font-size:17px;">2</div>
  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">サンプルデータを作る</div><div style='color:#555; margin-top:6px; line-height:1.7;'>3 店舗の日次売上（2025 年 4 月〜2026 年 9 月）を作ります。<b>異常を 3 日分わざと入れて</b>あります: 新宿店 7/20（近隣イベントで急増）、梅田店 8/15（システム障害で急減）、博多店 9/2（台風で半日休業）。</div></div>

```sql
CREATE OR REPLACE TABLE daily_sales
COMMENT '店舗別の日次売上（AI Functions デモ用の架空データ。2025年4月〜2026年9月。右肩上がりで年末に山がある。異常値を意図的に3日分入れてある）'
AS
WITH d AS (
  SELECT explode(sequence(DATE'2025-04-01', DATE'2026-09-30')) AS sales_date
), s AS (
  SELECT * FROM VALUES ('新宿店', 1.00D), ('梅田店', 0.80D), ('博多店', 0.55D) AS s(store, scale)
), base AS (
  SELECT
    sales_date, store, scale,
    datediff(sales_date, DATE'2025-04-01') AS t,
    CASE WHEN dayofweek(sales_date) IN (1, 7) THEN 1.35D ELSE 1.0D END AS weekend,
    -- 季節の波: 年末（12/24 ごろ）に大きな山、夏（8/10 ごろ）に小さな山
    1.0D
      + 0.32D * exp(-pow(least(abs(dayofyear(sales_date) - 358), 365 - abs(dayofyear(sales_date) - 358)), 2) / (2 * 14 * 14))
      + 0.08D * exp(-pow(dayofyear(sales_date) - 222, 2) / (2 * 18 * 18)) AS season,
    (abs(hash(store, sales_date)) % 1000) / 1000.0D AS r
  FROM d CROSS JOIN s
)
SELECT
  sales_date,
  store,
  CAST(ROUND(
    2000000 * scale * weekend * season * (1 + t * 0.0015D) * (0.93D + r * 0.14D)
    * CASE
        WHEN store = '梅田店' AND sales_date = DATE'2026-08-15' THEN 0.35D   -- システム障害でレジ停止
        WHEN store = '新宿店' AND sales_date = DATE'2026-07-20' THEN 1.90D   -- 近隣イベントで急増
        WHEN store = '博多店' AND sales_date = DATE'2026-09-02' THEN 0.40D   -- 台風で臨時休業（半日）
        ELSE 1.0D
      END, -3) AS BIGINT) AS sales_amount,
  CAST(ROUND(480 * scale * weekend * (0.9D + r * 0.2D)) AS INT) AS visitors
FROM base
ORDER BY store, sales_date;

ALTER TABLE daily_sales ALTER COLUMN sales_date COMMENT '営業日';
ALTER TABLE daily_sales ALTER COLUMN store COMMENT '店舗名';
ALTER TABLE daily_sales ALTER COLUMN sales_amount COMMENT '1日の売上金額（円）';
ALTER TABLE daily_sales ALTER COLUMN visitors COMMENT '1日の来店客数（人）';

SELECT * FROM daily_sales;
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">店舗ごとに異常な日を探す</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>group_col</code> に店舗を指定すると、店舗ごとに「いつも」を学んで判定します。ここでは直近の半年（2026 年 4〜9 月）を対象にします。異常と判定された行だけを表示します。</div></div>

```sql
SELECT
  sales_date, store, sales_amount,
  round(expected) AS expected,
  round(lower_bound) AS lower_bound,
  round(upper_bound) AS upper_bound,
  round(anomaly_score, 3) AS anomaly_score
FROM ai_detect_anomalies(
  TABLE(SELECT sales_date, store, sales_amount FROM daily_sales WHERE sales_date >= DATE'2026-04-01'),
  time_col       => 'sales_date',
  value_col      => 'sales_amount',
  group_col      => 'store',
  holiday_region => 'JP'
)
WHERE is_anomaly
ORDER BY sales_date
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">厳しさを変える</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>anomaly_threshold</code> を下げると、より小さな変化も拾います。<code>smoothing => false</code> にすると、異常の前後をつなげずに1日ずつ判定します。</div></div>

```sql
SELECT store, count_if(is_anomaly) AS anomaly_days
FROM ai_detect_anomalies(
  TABLE(SELECT sales_date, store, sales_amount FROM daily_sales WHERE sales_date >= DATE'2026-04-01'),
  time_col          => 'sales_date',
  value_col         => 'sales_amount',
  group_col         => 'store',
  anomaly_threshold => 0.8,
  smoothing         => false
)
GROUP BY store
ORDER BY store
```

```sql
CREATE OR REPLACE TABLE sales_anomalies
COMMENT 'daily_sales を ai_detect_anomalies で判定した結果（2026年4〜9月の全日分）'
AS
SELECT * FROM ai_detect_anomalies(
  TABLE(SELECT sales_date, store, sales_amount FROM daily_sales WHERE sales_date >= DATE'2026-04-01'),
  time_col => 'sales_date', value_col => 'sales_amount', group_col => 'store', holiday_region => 'JP');

ALTER TABLE sales_anomalies ALTER COLUMN expected      COMMENT 'いつもなら出ていたと推定される売上（円）';
ALTER TABLE sales_anomalies ALTER COLUMN anomaly_score COMMENT '異常の度合い（0〜1）';
ALTER TABLE sales_anomalies ALTER COLUMN is_anomaly    COMMENT '異常と判定されたか';

SELECT * FROM sales_anomalies WHERE is_anomaly ORDER BY sales_date;
```

      <div style="margin: 6px 0 0 0; color: #333; line-height:1.7;">2026 年 4〜9 月の、チャネル別・日別の配信数と配信停止の数です。<b>いつもと違う日を 3 日分わざと入れて</b>あります: メール 7/8（同じメールを2回送ってしまい、配信停止が急増）、LINE 8/20（配信の障害で途中で止まり、配信数が急減）、アプリ通知 9/10（深夜に通知を送ってしまい、配信停止が急増）。</div>

```sql
CREATE OR REPLACE TABLE message_metrics
COMMENT 'チャネル別・日別の配信数と配信停止の数（AI Functions デモ用の架空データ。2026年4〜9月。いつもと違う日を意図的に3日分入れてある）'
AS
WITH d AS (
  SELECT explode(sequence(DATE'2026-04-01', DATE'2026-09-30')) AS send_date
), c AS (
  SELECT * FROM VALUES ('メール', 120000, 0.0008D), ('LINE', 80000, 0.0012D), ('アプリ通知', 50000, 0.0015D) AS c(channel, base, optout_rate)
), b AS (
  SELECT
    send_date, channel, base, optout_rate,
    datediff(send_date, DATE'2026-04-01') AS t,
    CASE WHEN dayofweek(send_date) IN (1, 7) THEN 0.8D ELSE 1.0D END AS weekend,
    (abs(hash(channel, send_date, 'v')) % 1000) / 1000.0D AS r1,
    (abs(hash(channel, send_date, 'o')) % 1000) / 1000.0D AS r2
  FROM d CROSS JOIN c
), m AS (
  SELECT
    send_date, channel, optout_rate, r2,
    CAST(ROUND(base * weekend * (1 + t * 0.001D) * (0.95D + r1 * 0.1D)
      * CASE WHEN channel = 'LINE' AND send_date = DATE'2026-08-20' THEN 0.35D ELSE 1.0D END) AS INT) AS delivered   -- 配信の障害で途中で止まった
  FROM b
)
SELECT
  send_date, channel, delivered,
  CAST(ROUND(delivered * optout_rate * (0.8D + r2 * 0.4D)
    * CASE WHEN channel = 'メール' AND send_date = DATE'2026-07-08' THEN 6.0D        -- 同じメールを2回送ってしまった
           WHEN channel = 'アプリ通知' AND send_date = DATE'2026-09-10' THEN 4.0D   -- 深夜に通知を送ってしまった
           ELSE 1.0D END) AS INT) AS optouts
FROM m
ORDER BY channel, send_date;

ALTER TABLE message_metrics ALTER COLUMN send_date COMMENT '配信日';
ALTER TABLE message_metrics ALTER COLUMN channel COMMENT '配信チャネル（メール / LINE / アプリ通知）';
ALTER TABLE message_metrics ALTER COLUMN delivered COMMENT '届いた配信の数';
ALTER TABLE message_metrics ALTER COLUMN optouts COMMENT '配信停止（オプトアウト）の数';

SELECT * FROM message_metrics;
```

```sql
SELECT
  send_date, channel, metric, value,
  round(expected)        AS expected,
  round(lower_bound)     AS lower_bound,
  round(upper_bound)     AS upper_bound,
  round(anomaly_score, 3) AS anomaly_score
FROM ai_detect_anomalies(
  -- 配信数と配信停止の数を、縦に並べて1つの列（value）にする
  TABLE(SELECT send_date, channel,
               stack(2, '配信数', CAST(delivered AS DOUBLE), '配信停止', CAST(optouts AS DOUBLE)) AS (metric, value)
        FROM message_metrics),
  time_col  => 'send_date',
  value_col => 'value',
  group_col => ARRAY('channel', 'metric')   -- チャネル × 指標 ごとに「いつも」を学ぶ
)
WHERE is_anomaly
ORDER BY send_date
```

```sql
WITH parsed AS (
  SELECT path, ai_parse_document(content, map('version', '2.0')) AS doc
  FROM read_files(vol_root || '/docs/', format => 'binaryFile')
)
SELECT
  e.pos                                   AS element_no,
  e.value:bbox[0].page_id::int + 1        AS page,
  e.value:type::string                    AS type,
  e.value:content::string                 AS content,
  e.value:description::string             AS description
FROM parsed,
LATERAL variant_explode(parsed.doc:document.elements) AS e
ORDER BY element_no
```

```sql
CREATE OR REPLACE TABLE vendor_master
COMMENT '取引先マスタ（AI Functions デモ用の架空データ。請求書の発行元の確かめに使う）'
AS
SELECT * FROM VALUES
  ('株式会社ミナト物流',                   'V001'),
  ('グリーンオフィス株式会社',             'V002'),
  ('クラウドワークス・システムズ合同会社', 'V003')
AS t(vendor_name, vendor_code);

ALTER TABLE vendor_master ALTER COLUMN vendor_name COMMENT '取引先の会社名';
ALTER TABLE vendor_master ALTER COLUMN vendor_code COMMENT '取引先コード';

SELECT * FROM vendor_master;
```

```sql
WITH parsed AS (
  SELECT path, ai_parse_document(content, map('version', '2.0')) AS doc
  FROM read_files(vol_root || '/invoices/', format => 'binaryFile')
),
extracted AS (
  -- ① まず、組み込みの ai_extract で全件を読む
  SELECT
    element_at(split(path, '/'), -1) AS file_name,
    doc,
    ai_extract(
      doc,
      '{"vendor": {"type": "string", "description": "請求書の発行元の会社名（宛先の「御中」の会社ではない）"}, "total_jpy": {"type": "integer", "description": "税込の請求金額（円）"}, "line_items": {"type": "array", "items": {"type": "object", "properties": {"item": {"type": "string"}, "quantity": {"type": "integer"}, "unit_price": {"type": "integer"}}}}}',
      map('version', '2.1', 'mode', 'precision',
          'instructions', '日本の請求書から、請求した側（発行元）の情報を取り出します。請求書の上部で「御中」が付いている会社は宛先（請求された側）なので、vendor にしないでください。')
    ):response AS r
  FROM parsed
),
checked AS (
  SELECT
    file_name, doc,
    r:vendor.value::string    AS vendor,
    r:total_jpy.value::bigint AS total_jpy,
    aggregate(from_json(to_json(r:line_items), 'ARRAY<STRUCT<quantity: STRUCT<value: BIGINT>, unit_price: STRUCT<value: BIGINT>>>'),
              0L, (acc, x) -> acc + x.quantity.value * x.unit_price.value) AS subtotal
  FROM extracted
),
judged AS (
  -- ② SQL のルールで確かめる（LLM は呼ばない）
  SELECT
    c.*,
    c.vendor IS NOT NULL AND c.total_jpy IS NOT NULL AS rule_required,       -- 必須の項目が空でない
    round(c.subtotal * 1.1) = c.total_jpy             AS rule_total,          -- 明細の合計 × 1.1 ＝ 請求金額
    m.vendor_name IS NOT NULL                         AS rule_vendor_master   -- 発行元が取引先マスタにある
  FROM checked AS c
  LEFT JOIN vendor_master AS m ON c.vendor = m.vendor_name
),
reread AS (
  -- ③ ルールに引っかかった行だけを、ai_query で日本語に強いモデルに読み直させる
  SELECT
    file_name,
    ai_query(
      llm_endpoint,
      concat('次は請求書を読み取った結果です。請求した側（発行元）の会社名と、税込の請求金額を取り出してください。上部で「御中」が付いている会社は宛先なので、発行元にしないでください。

',
             to_json(doc:document.elements)),
      responseFormat => 'STRUCT<result: STRUCT<vendor: STRING, total_jpy: BIGINT>>'
    ) AS reread_result
  FROM judged
  WHERE NOT (rule_required AND rule_total AND rule_vendor_master)
)
SELECT
  j.file_name, j.vendor, j.total_jpy, j.subtotal,
  j.rule_required, j.rule_total, j.rule_vendor_master,
  j.rule_required AND j.rule_total AND j.rule_vendor_master AS passed,
  r.reread_result                                          -- ルールを通った行は ai_query を呼ばないので NULL
FROM judged AS j
LEFT JOIN reread AS r USING (file_name)
ORDER BY j.file_name
```

```sql
SELECT
  e.value:type::string    AS type,
  e.value:content::string AS content
FROM (
  SELECT ai_parse_document(content, map('version', '2.0')) AS doc
  FROM read_files(vol_root || '/images/', format => 'binaryFile')
) AS p,
LATERAL variant_explode(p.doc:document.elements) AS e
```

```sql
SELECT
  e.value:type::string    AS type,
  e.value:content::string AS content
FROM (
  SELECT ai_parse_document(content, map('version', '2.0', 'pageRange', '2', 'descriptionElementTypes', '')) AS doc
  FROM read_files(vol_root || '/docs/', format => 'binaryFile')
) AS p,
LATERAL variant_explode(p.doc:document.elements) AS e
```

```sql
CREATE OR REPLACE TABLE documents_parsed
COMMENT 'ボリュームの PDF・画像を ai_parse_document で読み取った結果'
AS
SELECT
  path,
  ai_parse_document(content, map('version', '2.0')) AS parsed
FROM read_files(vol_root || '/', format => 'binaryFile', recursiveFileLookup => 'true')
WHERE path RLIKE '\\.(pdf|png|jpg|jpeg)$';

ALTER TABLE documents_parsed ALTER COLUMN path   COMMENT '元のファイルのパス';
ALTER TABLE documents_parsed ALTER COLUMN parsed COMMENT 'ai_parse_document の戻り値（VARIANT）。document.elements に要素の配列';

SELECT path, parsed:error_status AS error_status FROM documents_parsed;
```

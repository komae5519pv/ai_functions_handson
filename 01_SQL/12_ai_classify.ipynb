```sql
SELECT
  inquiry_id,
  body,
  ai_classify(
    body,
    '{
      "配送":             "商品が届かない、遅れている、お届け日の変更",
      "支払い・請求":     "請求金額、支払い方法、見積との差異",
      "製品の故障・安全": "動かない、壊れた、発熱や異臭など安全に関わること",
      "アカウント":       "ログイン、パスワード、会員情報の変更",
      "返品・交換":       "返品したい、交換したい",
      "購入の相談":       "購入前の質問、まとめ買い、見積の依頼",
      "その他":           "上のどれにも当てはまらないもの（お礼や感想など）"
    }',
    map('version', '2.1',
        'instructions', '家電・日用品を扱う EC サイトのサポート窓口に届いた問い合わせです。',
        'enableConfidenceScores', 'true',
        'enableRationales', 'true')
  ) AS result
FROM inquiries
```

```sql
WITH classified AS (
  SELECT
    inquiry_id,
    body,
    ai_classify(
      body,
      '["配送", "支払い・請求", "製品の故障・安全", "アカウント", "返品・交換", "購入の相談", "その他"]',
      map('version', '2.1', 'enableConfidenceScores', 'true')
    ) AS result
  FROM inquiries
)
SELECT
  inquiry_id,
  body,
  result:response[0].value::string            AS category,
  result:response[0].confidence_score::double AS confidence,
  result:error_message::string                AS error_message
FROM classified
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">1つの文章に複数のラベルを付ける</div><div style='color:#555; margin-top:6px; line-height:1.7;'><code>'multilabel','true'</code> を付けると、当てはまるラベルがすべて返ります。話題が複数ある文章に向いています。</div></div>
</div>

```sql
SELECT
  inquiry_id,
  body,
  transform(
    cast(ai_classify(
      body,
      '["急ぎの対応が必要", "お金に関わる", "安全に関わる", "個人情報を含む", "前向きな声"]',
      map('version', '2.1', 'multilabel', 'true')
    ):response AS ARRAY<VARIANT>),
    x -> x:value::string
  ) AS tags
FROM inquiries
```

  <div style="flex:1;"><div style="font-size:20px; font-weight:700; color:#1a2744; line-height:36px;">結果をテーブルに保存する</div><div style='color:#555; margin-top:6px; line-height:1.7;'>分類結果は保存しておき、担当部署への振り分けや件数の集計はこのテーブルから行います。</div></div>

```sql
CREATE OR REPLACE TABLE inquiries_classified
COMMENT 'inquiries に ai_classify の分類結果を付けたテーブル'
AS
SELECT
  *,
  ai_classify(
    body,
    '["配送", "支払い・請求", "製品の故障・安全", "アカウント", "返品・交換", "購入の相談", "その他"]',
    map('version', '2.1', 'enableConfidenceScores', 'true')
  ) AS result
FROM inquiries;

ALTER TABLE inquiries_classified ALTER COLUMN result COMMENT 'ai_classify の戻り値（VARIANT）。response[0].value が分類ラベル';

SELECT
  result:response[0].value::string AS category,
  count(*) AS inquiry_count
FROM inquiries_classified
GROUP BY ALL
ORDER BY inquiry_count DESC;
```
